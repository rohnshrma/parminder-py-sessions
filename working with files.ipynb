# Working with Files in Python

So far every value we created (variables, lists, dictionaries) lived only in memory - the moment the program stops, it's gone.
**Files** let us save data on the disk so it stays there even after the program ends, and read it back later.

Every file operation follows the same 3 steps:

1. **Open** the file with `open(path, mode)` - this gives us a *file object* to work with
2. **Read from it / write to it** using the file object's methods
3. **Close** it with `.close()` - this saves everything properly and frees the file

In this notebook our files live in the `DATA` folder, next to this notebook.

## 1. Reading the whole file - `read()`
`read()` reads **everything** in the file and gives it back as **one single string** (new lines included as `\n`).

```python
# open() takes 2 things: the path of the file and the mode
# "./DATA/data.txt" -> "." means the current folder, then go inside DATA and pick data.txt
# "r" means read mode - we only want to look at the file, not change it
# open() gives back a file object, which we store in the variable file
file = open("./DATA/data.txt" , "r")

# read() pulls out the entire content of the file as one string
data = file.read()

# print the string - both lines show because the "\n" inside the string moves to the next line
print(data)

# always close the file once you're done with it
file.close()
```

## File Modes
The second value we give to `open()` is the **mode**. It tells Python *what* we want to do with the file.

| Mode | Meaning | If file doesn't exist | What happens to old content |
|------|---------|------------------------|-----------------------------|
| `r`  | Read only (default) | Error | Untouched |
| `w`  | Write only | Creates it | **Erased** (file starts empty) |
| `a`  | Append (add at the end) | Creates it | Kept, new data goes at the end |
| `x`  | Create a new file | Creates it | Error if file already exists |
| `r+` | Read + Write | Error | Kept |
| `w+` | Write + Read | Creates it | **Erased** |
| `a+` | Append + Read | Creates it | Kept, writing goes at the end |

The three we use the most are **`r`**, **`w`** and **`a`**.

## 2. Reading one line - `readline()`
`readline()` reads only **one line** each time it's called. The next call gives the next line, and so on.

```python
# open the same file again in read mode
file = open("./DATA/data.txt" , "r")

# readline() gives only the FIRST line (including the "\n" at its end)
# if we called file.readline() again, we would get the second line
data = file.readline()

# notice the extra empty line in the output:
# the line already ends with "\n" and print() adds one more new line of its own
print(data)

file.close()
```

## 3. Reading all lines into a list - `readlines()`
`readlines()` reads the whole file but gives back a **list**, where **each line is one item** of the list.

```python
file = open("./DATA/data.txt" , "r")

# readlines() -> a list of strings, one string per line
# the "\n" stays at the end of every line except the last one
# since it's a list, we can use indexing, len(), loops etc. on it
data = file.readlines()

print(data)

file.close()
```

## 4. Looping over a file line by line
A file object can be used directly in a `for` loop - each round of the loop gives us the next line.

```python
file = open("./DATA/data.txt" , "r")

# the loop variable "line" holds one line of the file in each round
for line in file:
    # strip() is the string method we learned earlier - it removes spaces and "\n" from both ends
    # without strip() we would get an empty line after every line (same issue as readline above)
    print(line.strip())

file.close()
```

### Quick comparison of the reading methods

| Method | Gives back | Use it when |
|--------|-----------|-------------|
| `read()` | one string (whole file) | you want all the text at once |
| `readline()` | one string (one line) | you need just the next line |
| `readlines()` | a list of strings | you want to use list tools on the lines |
| `for line in file` | one line per loop round | you want to process each line one by one |

## 5. Writing to a file - `"w"` mode and `write()`
- Opening in `"w"` mode **creates** the file if it isn't there.
- If the file already exists, `"w"` **wipes out everything** in it first.
- `write()` does **not** add a new line on its own - we have to put `\n` ourselves.

```python
# "w" = write mode -> student.txt is created (or emptied if it already exists)
file = open("./DATA/student.txt", "w")

# write() puts the string into the file exactly as it is
# the "\n" at the end makes each name go on its own line
file.write("Parminder\n")
file.write("John\n")
file.write("Rohan\n")

# closing saves what we wrote to the disk
file.close()
```

```python
# opening the SAME file in "w" mode again...
# ...erases Parminder, John and Rohan first! Only the three new names will remain
file = open("./DATA/student.txt", "w")

file.write("Josie\n")
file.write("Jack\n")
file.write("Johnny\n")
file.close()
```

## 6. Writing many lines at once - `writelines()`
`writelines()` takes a **list of strings** and writes all of them into the file one after another.

```python
# a normal list of names - each string already has "\n" at the end
# because writelines() also does NOT add new lines on its own
lines = ["Rohan\n","Josie\n","Parminder\n"]
```

```python
# "w" again -> old content is erased first
file = open("./DATA/student.txt", "w")

# writelines() loops through the list for us and writes every string into the file
file.writelines(lines)

file.close()
```

## 7. Adding to a file - `"a"` (append) mode
`"a"` mode **keeps the old content** and adds the new data **at the end** of the file.
This is the difference between `"w"` (replace everything) and `"a"` (add more).

```python
# "a" = append mode -> Rohan, Josie, Parminder stay in the file
file = open("./DATA/student.txt", "a")

# these 3 names get added below the existing names
file.write("Josie\n")
file.write("Jack\n")
file.write("Johnny\n")
file.close()

# the file now has 6 lines: Rohan, Josie, Parminder, Josie, Jack, Johnny
```

## 8. Mini Project - File Manager
Here we combine everything: **functions**, a **`while True`** loop for the menu, **`input()`**, **`if / elif / else`**, **`continue`**, **`break`** and the **three file modes** (`r`, `w`, `a`).

```python
# a function that takes the path of the file we want to manage
# because the path is a parameter, the same function works for ANY file
def fileManager(path):
    print("==== FILE MANAGER====")

    # while True = an endless loop, so the menu keeps coming back after every task
    # the only way out is the "break" in option 4
    while True:
        # show the menu options
        print("\n==== Menu ====")
        print("1. Read File")
        print("2. Write File")
        print("3. Add Into File")
        print("4. Exit")

        # input() always gives a string, so int() turns it into a number we can compare
        op = int(input("\nEnter your choice (1-4) :"))

        # if the number is not between 1 and 4, tell the user...
        if  op < 1 or op > 4:
            print("Invalid Operation")
            # ...and continue skips the rest of this round and shows the menu again
            continue

        if op == 1:
            # option 1: read mode -> read the full file and show it
            file = open(path, "r")
            data = file.read()
            # f-string to put the data inside the message
            print(f"DATA => \n{data}")
            file.close()
        elif op == 2:
            # option 2: write mode -> everything already in the file is erased
            file = open(path, "w")
            data = input("enter data to overwrite : ")
            # add "\n" so the next thing written starts on a new line
            file.write(data + "\n")
            print("File OverWritten Successfully")
            file.close()
        elif op == 3:
            # option 3: append mode -> old content stays, new data goes at the end
            file = open(path, "a")
            data = input("enter data to add : ")
            file.write(data + "\n")
            print("File Written Successfully")
            file.close()
        else :
            # the only number left is 4 -> say bye and break out of the while loop
            print("Thanks for using file manager!")
            break
```

```python
# calling the function with the path of our student file
# try: 1 (read), 2 (overwrite), 1 (read again), 3 (add), 1 (read again), 4 (exit)
# watch how option 2 replaces everything but option 3 only adds to the end
fileManager("./DATA/student.txt")
```

## Key Points to Remember
- `open(path, mode)` -> gives a file object; always `close()` it when done.
- `r` reads, `w` **erases then writes**, `a` **adds at the end**.
- `read()` -> string, `readline()` -> one line, `readlines()` -> list of lines.
- `write()` and `writelines()` never add `\n` for you - add it yourself.
- Lines read from a file keep their `\n` - use `strip()` to clean them.
- Everything read from a file is a **string**. To do maths with it, convert it with `int()` / `float()`.
