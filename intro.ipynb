## What is a comment?

A **comment** is text that Python ignores while it runs a program. Comments explain code, leave reminders, or temporarily disable a line. A single-line comment begins with `#`.

```python
# Python ignores this entire line.
print("This line runs")  # This comment explains the output.
```

Good comments explain *why* something is done or clarify a new concept. They should support the code rather than repeat it word for word.

## 1. Output with `print()`

````python
# print() sends a value to the output area.
print(23 ** 2)  # ** raises 23 to the power of 2.
````

## 3. Variables and assignment

A **variable** is a name that refers to a value. The assignment operator `=` stores the value on its right under the name on its left.

```text
age = 23
name = "Alex"
```

Python creates a variable when the assignment statement runs. It does not require a separate declaration first.

````python
# The name 'age' now refers to the integer value 23.
age = 23

# Assignment stores a value; it does not mean mathematical equality.
print("Age:", age)
````

````python
# Variables can be used in expressions.
age_in_two_years = age + 2
print("Age in two years:", age_in_two_years)
````

## 4. Variable naming conventions

### Rules

- Use letters, digits, and underscores only.
- Do not use spaces.
- Do not begin with a digit.
- Python is case-sensitive: `age` and `Age` are different names.
- Do not use a reserved keyword as a variable name.

### Style suggestions

Use descriptive lowercase `snake_case` names for multiword variables:

- `student_name` is clear and readable.
- `studentName` is valid, but `snake_case` is the usual Python style.
- `studentname` is valid, but less readable.

The following examples are shown as comments because they would each raise a `SyntaxError` if executed:

```python
# my age = 23       # A space is not allowed.
# 12age = 43        # A name cannot begin with a digit.
# $%^age = 24      # These symbols are not allowed.
# for = 34          # 'for' is a reserved keyword.
```

Underscores are allowed, but excessive underscores make names harder to read. A name such as `total_age` is preferable to `___________age________`.

### Python reserved keywords

Reserved keywords have a special meaning in Python and cannot be used as ordinary variable names.

| Keyword | Purpose | Keyword | Purpose |

| `case` | Pattern-match case | `class` | Define a class |
| `continue` | Skip to next loop step | `def` | Define a function |
| `del` | Delete a reference | `elif` | Else-if branch |
| `else` | Alternative branch | `except` | Handle an exception |
| `False` | Boolean false | `finally` | Always-run cleanup |
| `for` | Iterate over items | `from` | Import from a module |
| `global` | Use a global name | `if` | Conditional branch |
| `import` | Load a module | `in` | Test membership |
| `is` | Test object identity | `lambda` | Create an anonymous function |
| `match` | Start pattern matching | `None` | No value |
| `nonlocal` | Use an enclosing name | `not` | Logical NOT |
| `or` | Logical OR | `pass` | Do nothing placeholder |
| `raise` | Trigger an exception | `return` | Send a function result |
| `True` | Boolean true | `try` | Start exception handling |
| `while` | Repeat while true | `with` | Manage a context |
| `yield` | Produce a generator value |  |  |

Python also has soft keywords such as `_` and `type` in specific contexts. The table above lists the standard reserved keywords used by modern Python.

## 5. Python data types

A data type describes what kind of value a variable holds. Python determines the type automatically when a value is assigned.

| Type | Meaning | Example |
| --- | --- | --- |
| `int` | Whole number | `23` |
| `float` | Number with a decimal part | `3.14` |
| `bool` | A truth value | `True` or `False` |
| `str` | Text inside quotes | `"hello"` |
| `NoneType` | Absence of a value | `None` |

Use `type(value)` to inspect a value's type. A string from `input()` can be converted to a number with `int()` or `float()` when appropriate.

````python
# Python creates the variable and infers its type from the value.
age = 23
height = 1.75
is_learning = True
name = "Alex"

print(type(age))
print(type(height))
print(type(is_learning))
print(type(name))
````

````python
# None means that a value is intentionally absent or not known yet.
email = None
print("Email:", email)
print("Type:", type(email))
````

````python
# Strings are sequences of characters enclosed in quotes.
name = "Alex"
print(name)
print(type(name))
````
