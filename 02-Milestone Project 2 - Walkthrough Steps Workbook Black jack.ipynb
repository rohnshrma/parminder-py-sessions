## Game Play

Here is what happens in one hand of Blackjack. Read it through once now, then come back to it whenever you get lost, because every step in this notebook maps to one or more of these lines.

1. Create a deck of 52 cards
2. Shuffle the deck
3. Ask the Player for their bet
4. Make sure that the Player's bet does not exceed their available chips
5. Deal two cards to the Dealer and two cards to the Player
6. Show only one of the Dealer's cards, the other remains hidden
7. Show both of the Player's cards
8. Ask the Player if they wish to Hit, and take another card
9. If the Player's hand doesn't Bust (go over 21), ask if they'd like to Hit again.
10. If a Player Stands, play the Dealer's hand. The dealer will always Hit until the Dealer's value meets or exceeds 17
11. Determine the winner and adjust the Player's chips accordingly
12. Ask the Player if they'd like to play again

**Words you need to know**
* **Hit**: take one more card.
* **Stand**: stop taking cards and keep what you have.
* **Bust**: your total went over 21, so you lose right away.
* **Push**: you and the dealer tie. Nobody wins or loses chips.

The goal is to get closer to 21 than the dealer without going over.

## Playing Cards

A standard deck has four suits (Hearts, Diamonds, Spades and Clubs) and thirteen ranks (2 through 10, then Jack, Queen, King and Ace), so 52 cards in total.

Card values in Blackjack are different from what you might expect:
* Number cards are worth their number.
* Jack, Queen and King are all worth 10.
* An Ace is worth **11 or 1**, whichever helps the hand more. This is the trickiest part of the whole project, so keep it in the back of your mind. A hand of Ace + Nine is 20, but if you hit and get a Five, the Ace has to turn into a 1 or you would bust with 25.

To store all this, use a tuple for the suits, a tuple for the ranks, and a dictionary that connects each rank to its value. You will look values up in that dictionary a lot.
