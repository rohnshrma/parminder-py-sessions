## Simple War Game

Before the Milestone 2 project, we're going to build something a bit bigger than the usual toy class: the card game War. It's a good way to see why OOP is useful, because the game is basically three things (cards, a deck, and players) and each one is its own class.

The rules are simple:
* Two players each get half of a shuffled deck (26 cards each).
* Every round, both players put down their top card.
* The higher card wins, and that player takes both cards and puts them at the bottom of their pile.
* If the cards are the same value, it's a **war**. Each player puts down 5 more cards face down and then the last card put down decides who takes everything on the table.
* If a player runs out of cards, they lose.
