```python
def div(num):
    try:
        x = int(input("Enter Number 2 :"))
        res = num / x
    except ValueError:
        print("The input has be a number")
    except ZeroDivisionError:
        print("Failed! Cannot divide by zero")
    except:
        print("Oops! something went wrong")
    else:
        print(res)
    finally:
        print("Function Ends here....")
```
