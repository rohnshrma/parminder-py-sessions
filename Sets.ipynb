# Sets
# unordered, unindexed, unchangable, no duplicates

`{ }` or `set()` are used to create a set where items are seperated using `,`

```python
# { } creates a set - like a list, but it automatically throws away duplicates and doesn't care about order
marks = {100 , 200, 300 , 400 , 400}
```

```python
print(type(marks))   # confirms marks is a "set" object
print(len(marks))    # even though 400 was typed twice, sets auto-remove duplicates - so len() only counts 4 unique items
```

```python
print(marks) # unordered = python may print these in ANY order, not necessarily how you typed them; no duplicates survive
```

```python
marks[0] # unindexed - crashes on purpose: sets have NO position/order, so there's no "box 0" to ask for
```

```python
# this crashes too, for a related reason: there's no index [0] to point at, so there's nothing to assign to
# (side note: sets CAN still grow/shrink using .add() / .remove() / .discard() a few cells below -
# "unchangeable" here really just means you can't reach in and swap ONE specific item, since there's no index system)
marks[0] = 100000 # unchangable (immtable)
```

```python
# .add(item) puts a new item into the set (if it's already there, nothing happens - no duplicates allowed)
marks.add(1099)
print(marks)
```

```python
# .pop() removes and returns an ARBITRARY item - since sets have no order, there's no "last item" to target like lists have
marks.pop()
print(marks)
```

```python
# .remove(value) deletes that exact value from the set - but throws a KeyError if the value isn't there at all
marks.remove(300)   # 300 exists in the set, so this removes it fine
marks.remove(999)   # 999 was never in the set - this line crashes on purpose to show the error

print(marks)
```

```python
# .discard(value) does the same job as .remove(), but stays quiet (does nothing) if the value isn't found - no crash
marks.discard(300)   # already removed above, so this just does nothing
marks.discard(999)   # never existed in the first place, still no error either way

print(marks)
```

```python
# just a speed check: sets are built for super-fast lookups/membership tests, much faster than searching through a list
%timeit marks.discard(100)
```

```python
# .union(otherSet) combines both sets into a NEW set, keeping only the unique values from both (originals stay untouched)
x = marks.union({"rohan","sohan","mohan"})
print(x)
```

```python
# two sets with some overlap (3 and 4 are in both) - used below to demo set comparisons
a = {1,2,3,4}
b = {3,4,5,6}
```

```python
# .intersection() returns a NEW set with only the items that exist in BOTH sets
a.intersection(b)
```

```python
# same comparison as .intersection(), but this OVERWRITES "a" itself instead of just returning a new set
a.intersection_update(b) # a = {3,4}
```

```python
# .difference() returns a NEW set with items that are in "a" but NOT in "b"
a.difference(b)
```

```python
# same comparison as .difference(), but this OVERWRITES "a" itself instead of returning a new set
a.difference_update(b) # a = {1,2}
```

```python
# .symmetric_difference() returns a NEW set with everything that's in EITHER set but NOT in both (the opposite of intersection)
a.symmetric_difference(b)
```

```python
# same comparison as .symmetric_difference(), but this OVERWRITES "a" itself instead of returning a new set
a.symmetric_difference_update(b)
```

```python
a   # confirm that "a" really was overwritten by symmetric_difference_update above
```

```python

# fresh sets for the subset/superset checks below - notice everything in "a" also happens to be inside "b"
a = {1,2,3,4}
b = {1,2,3,4,5,6,7}
```

```python
a.issubset(b)   # True if EVERY item in "a" also exists somewhere in "b" (a fits entirely inside b)
```

```python
b.issuperset(a)   # True if "b" contains EVERY item that's in "a" - the reverse check of issubset
```

```python
# a completely separate set of names, with nothing in common with "x" (from the union example earlier) - used below
names = {"John", "Josie", "Jackie"}
```

```python
# .isdisjoint(otherSet) checks if two sets have ZERO items in common - True means they share absolutely nothing
print(names.isdisjoint(x))
```

```python
x   # reminder of what "x" holds, from the .union() example earlier in the notebook
```

```python
names   # confirming names never shared anything with x, which is why isdisjoint() came back True above
```

```python
# .copy() makes an independent copy, same idea as list.copy() - changing one won't affect the other
n = names.copy()
```

```python
n   # a separate copy holding the same 3 names
```

```python
# .clear() empties the set completely
n.clear()
```

```python
n   # python shows an empty set as set(), NOT as {} - because {} on its own actually means an empty dictionary, not a set
```
