# Collections:
collections are used to store multiple values of similar or different data types

- List : ordered, indexed, changable (mutable), Allows Duplicates
- Tuple : ordered, indexed, unchangable (immutable), allows duplicates
- Dictionary : ordered, unindexed, changable(mutable), no duplicates are allowed
- Set : unordered, unindexed, unchangable (immutable), no duplicates are allowed

# List
[ ] are used to create a list where items are divided using "," 

 # ordered, indexed, changable (mutable), Allows Duplicates

```python
# [ ] creates a list - a container that holds multiple values together, in the exact order you typed them
marks = [23,12,34,12,45]
```

```python

print(type(marks))   # confirms marks is a "list" object
```

```python
print(len(marks))   # len() works on lists too - counts how many ITEMS are inside, not characters -> 5 items
```

```python
print(marks)   # ordered = items stay in the exact sequence you put them in; duplicates allowed = 12 can appear twice, no problem
```

```python
# lists use the exact same indexing/slicing rules as strings - box 0 is first, -1 is last
print(marks[2])          # single item at position 2 -> 34
print(marks[2 : 4])      # slice: position 2 up to (not including) position 4 -> [34, 12]
print(marks[-3:-1])      # negative slicing works too - 3rd-from-last up to (not including) last -> [34, 12]
```

```python
# unlike strings, lists are "mutable" - you CAN change an item after the list already exists
marks[2] = 43   # overwrite whatever was sitting at position 2 with 43
print(marks)
```

```python
# List Methods
# just like strings had .upper(), .lower(), etc, lists come with their own built-in toolbox of methods
```

```python
# .append(item) adds ONE new item to the very END of the list (the list grows by exactly 1)
marks.append(100)
print(marks)
```

```python
# .insert(position, value) squeezes a new value in at that exact position, pushing everything after it one step to the right
marks.insert(2 , 99) # insert position (inclusive) , value
print(marks)
```

```python
extras = [123,456,56,3]
# .extend(otherList) glues all the items from another list onto the end, one by one (NOT as one single nested list)
marks.extend(extras) #list of items
print(marks)
```

```python
# .pop() removes and RETURNS an item - with no argument, it removes the LAST item by default
marks.pop()  # removes the very last item (3)
marks.pop(6) # giving a position instead removes whatever item is sitting at that exact index (100)
print(marks)
```

```python
# .remove(value) searches for that VALUE (not a position) and deletes the first one it finds
marks.remove(12)
print(marks)
```

```python
# .count(value) tells you how many times that value shows up in the list
print(marks.count(3))
```

```python
# .index(value) tells you the position of the first time that value appears (raises an error if it isn't in the list at all)
print(marks.index(3)) # item
print(marks.index(3 , 9)) # item , start position (inclusive)
# this line crashes on purpose - there's no "3" anywhere from position 9 to the end of the list
```

```python
# .reverse() flips the order of the list IN PLACE - it permanently rewrites "marks" itself
marks.reverse() 
print(marks)
```

```python
# note: [::-1] slicing gives back a REVERSED COPY without touching the original - different from .reverse() above,
# which actually rewrote "marks" itself for good. this line is just a one-time look, marks doesn't change because of it
marks[::-1] # temp
```

```python
# .sort() rearranges the list IN PLACE, smallest to largest by default
marks.sort() # a-z
print(marks)
marks.sort(reverse=True) # reverse=True flips it to largest-to-smallest instead
print(marks)
```

```python
# .copy() makes a completely separate, independent copy of the list (why this matters gets shown a few cells down)
yourMarks = marks.copy()
```

```python
yourMarks   # peek at the copy - same values, but a totally separate list in memory
```

```python
# .clear() empties the list completely, leaving it as [] - the variable still exists, it's just empty now
yourMarks.clear()
print(yourMarks)
```

```python
# del completely deletes the VARIABLE itself (not just its contents) - after this "yourMarks" no longer exists at all
del yourMarks
```

```python
yourMarks   # this crashes with NameError - we just deleted the variable above, so python has no idea what "yourMarks" is anymore
```

```python

# quick detour before the list version: how does "=" behave for simple values like numbers?
a = 23
b = a # b gets its OWN separate copy of a's value (23) - they are equal for now, but not linked
print(a,b)
a = 100  # changing 'a' afterward has zero effect on 'b' - they were never connected, just equal at that one moment
print(a , b)
```

```python
# now the same idea, but with a LIST this time - and this is where it gets tricky
marks = [1,2,3]
yourMarks = marks   # this does NOT make a copy! both names now point to the exact same list sitting in memory
print(marks , yourMarks)
```

```python
marks[0] = 100
# changing marks also changes yourMarks - because they were never two separate lists, just two labels pointing at ONE list
print(marks,yourMarks)
```

```python
# .copy() is the fix - it creates a real, independent second list carrying the same values
myMarks = marks.copy()
print(myMarks , marks)
```

```python
marks[0] = 1000
# this time only "marks" changes - myMarks is a true separate copy, so it's completely unaffected
print(myMarks , marks)
```
