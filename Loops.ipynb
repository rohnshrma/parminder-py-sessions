# While Loop (Fixed Iteration)
A `while` loop is python's way of saying: "keep repeating this block of code, over and over,
for as long as some condition stays True." The instant the condition turns False, python skips
the loop entirely and moves on to whatever code comes after it.

Every while loop needs 3 ingredients. Miss one and the loop either never runs, or runs forever by mistake:
- **Initialization** — give your counter variable a starting value, done once, before the loop begins
- **Condition** — the True/False question python re-checks before every single repeat (each repeat is
  called an "iteration"); as long as the answer is True, the loop keeps going
- **Update** — something inside the loop that changes the counter a little each time (like adding 100),
  so that eventually the condition flips to False and the loop actually stops

If you forget the update step, the condition never changes, so it stays True forever — that's an
"infinite loop" and it will freeze your program (you'd have to force-stop it).

```python
i = 0 # initialization: start counting from 0
while i < 1000: # condition: keep looping as long as i is still less than 1000
    print(i)
    i += 100 # update: add 100 to i every time round, so it eventually reaches 1000 and the loop stops
```

````python
i = 0 # initialization: this is our counter, and it starts at 0

# condition: before every single loop, python checks "is i still less than 1000?"
# the moment this becomes False, the loop stops and python jumps to whatever code is after it
while i < 1000: # condition
    # update: we add 100 to i FIRST, before we print anything
    i += 100 # update
    print(i)

# because we update i before printing, the very first thing printed is 100, not 0
# and it keeps going: 200, 300, 400... all the way up to 1000
# once i actually becomes 1000, the condition "i < 1000" turns False, so the loop stops right there
# that's why 1000 is the last number printed, and the loop never prints anything past it
````

````python
# a list of numbers we want to go through one at a time and check
marks = [12,23,12,5,23,45,23,34]

# an empty list we'll use to collect only the even numbers we find inside marks
# it starts completely empty, the loop below is what fills it up
evens = []
````

````python
i = 0 # initialization: i is our position (index) in the list, we start at the first item, index 0

# condition: len(marks) tells us how many items are in the list (8 items here)
# so this keeps looping as long as i is still a valid position inside the list (0,1,2...7)
# once i reaches 8, the condition becomes False and the loop stops - this stops us from
# trying to look at a position that doesn't exist in the list ("index out of range")
while i < len(marks):
    # marks[i] grabs whatever number is sitting at position i in the list
    # % is the "remainder after division" operator - if a number % 2 leaves no remainder (0), it's even
    if marks[i] % 2 == 0:
        # .append() sticks this even number onto the end of the evens list
        evens.append(marks[i])
    # update: move on to the next position in the list
    # this line MUST be outside the "if" (but still inside the while), so it runs every single
    # time round the loop, even for the numbers that were odd and got skipped
    # if this line were missing, i would stay 0 forever and the loop would never stop
    i+=1

print(evens)

# walking through it: marks = [12,23,12,5,23,45,23,34]
# 12 is even (kept), 23 is odd (skipped), 12 is even (kept), 5 is odd (skipped),
# 23 is odd (skipped), 45 is odd (skipped), 23 is odd (skipped), 34 is even (kept)
# that's why evens ends up as [12, 12, 34]
````

````python
email = input("Enter your email : ")

while len(email) < 8 or not email.endswith(".com") or not "@" in email:
    print("\nInvalid Email!")
    email = input("\nEnter your email : ")

print("Valid Email : " , email)
````

````python
i = 1
while i <= 100:
    if i % 3 == 0 and i % 7 == 0:
        print("Im done! lets take a break")
        break
    print(i)
    i+=1
````

````python
i = 1
while i <= 100:
    if i % 3 == 0 and i % 7 == 0:
        print("Im done! lets skip")
        i+=1
        continue
    print(i)
    i+=1
````

````python
from random import randint
secret = randint(1,101)
attempts = 1

while attempts <= 10:
    attempts += 1
    guess = int(input("Enter your guess: "))
    
    if guess > 100 or guess < 1:
        print("Invalid Range")
        continue
    
    if guess > secret:
        print('Too high!')
    elif guess < secret:
        print('Too low!')
    else:
        print(f"Congrats! You've guesses the correct number {secret} in {attempts} attempts")
        break

if attempts > 10:
    print("Failed to guess! the correct number was :", secret)
````
