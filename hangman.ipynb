```python

# hangman is a list of strings, and each string is one "stage" of the drawing
# stage 0 is just the gallows with nothing hanging, stage 8 is the full body (game over)
# every time the player guesses wrong, we'll move one stage further into this list
# so len(hangman) tells us how many wrong guesses we can survive before the drawing is complete
hangman = [
"""
__________
""",
"""
__________
    |
""",
"""
__________
    |
    |

""",
"""
__________
    |
    |
    O

""",
"""
__________
    |
    |
    O
    |
    |

""",
"""
__________
    |
    |
    O
   /|
    |

""",
"""
__________
    |
    |
    O
   /|\\
    |

""",
"""
__________
    |
    |
    O
   /|\\
    |
   /
""",
"""
__________
    |
    |
    O
   /|\\
    |
   / \\
"""
]

```

```python
from random import choice # choice() picks one random item out of a list, that's how we pick the secret word

words = ["panda" , "dog", "apple", "dragonfruit", "papaya"] # the pool of possible secret words
word = choice(words) # word is now whichever one of those got picked randomly this run

# blanks is what the player actually sees instead of the real word - one underscore per letter
# ["_"] * len(word) repeats the single-item list ["_"] as many times as word has letters
# e.g. if word is "dog" (3 letters), blanks becomes ['_', '_', '_']
blanks = [ "_" ] * len(word)
print(blanks , word) # debug line - prints the real word too, handy while building/testing, but spoils the game if left in

guessedLetters = set() # a set to remember every letter already tried, so we can block duplicate guesses
                        # (a set is used instead of a list because checking "have i seen this before" is faster,
                        # and it automatically ignores duplicates)
wrongGuesses = [] # a list of only the WRONG letters guessed so far, in the order they happened
maxWrong = len(hangman) - 1 # how many wrong guesses end the game - one less than the number of drawing stages,
                             # because the first stage (empty gallows) doesn't need a wrong guess to appear

print("Welcome to Hangman")

# the game keeps looping as long as BOTH things are true:
# 1) there's still at least one "_" left in blanks (word isn't fully guessed yet)
# 2) we haven't run out of wrong guesses yet
# the moment either condition fails, the loop stops on its own
while "_" in blanks and len(wrongGuesses) < maxWrong:
    # " ".join(blanks) glues the list of letters/underscores into one string with spaces between them
    # e.g. ['_','_','a'] becomes "_ _ a" so it's easier to read on screen
    print(f"\nSECRET WORD: {" ".join(blanks)}")
    print(f"CHANCES REMAINING: {maxWrong - len(wrongGuesses)}")
    print(f"WRONG GUESSES: {", ".join(wrongGuesses)}\n")

    # .lower() makes the guess case-insensitive, so typing "A" counts the same as "a"
    guess = input("Guess a letter:\n").lower()

    # validation: guess must be exactly one alphabet character
    # .isalpha() rejects numbers/symbols/spaces, and len(guess) != 1 rejects whole words
    if not guess.isalpha() or len(guess) != 1:
        print("Guess Must be a single letter!")
        continue # continue skips everything below and jumps straight back to the while condition
                 # this means an invalid guess does NOT cost the player a turn or a wrong guess

    # if this exact letter was already tried before (right or wrong), don't let them guess it again
    if guess in guessedLetters:
        print(f"You already guessed '{guess}'. Try a different letter.")
        continue # same idea - skip the rest, go straight back to checking the while condition

    guessedLetters.add(guess) # now that it's passed both checks, record it as an attempted letter

    if guess in word: # check if the guessed letter appears anywhere in the secret word
        # enumerate(word) gives us both the position (index) and the letter at that position, one at a time
        # we need the index because blanks is a list, and we have to update the exact same position in it
        for index ,letter in enumerate(word):
            if letter == guess:
                blanks[index] = guess # reveal the letter at this position instead of "_"
        print(f"Good guess! The letter {guess} is in the secret word.")

    else:
        wrongGuesses.append(guess) # log this as one more wrong guess
        print(f"Oh no! The letter {guess} is not in the word.")
        # len(wrongGuesses) is now the count of wrong guesses so far, which we use as the index
        # into hangman to show the next, slightly-more-complete drawing
        print(hangman[len(wrongGuesses)])


# once the loop above ends, we're here to figure out WHY it ended - a win or a loss
if "_" not in blanks: # no underscores left means every letter in the word has been revealed
    print(f"\nSecret Word: {" ".join(blanks)}")
    print("Congratulations you won!")
else:
    # if we're in this branch, the loop must have stopped because wrongGuesses hit maxWrong
    print(f"\nYou Lose!\nSecret Word : {word}")
    print(f"Wrong guesses : {wrongGuesses}")
    print(hangman[maxWrong]) # show the final, fully-drawn hangman stage
```
