# Assignment: Employee Record System

**Real-life problem:** A small company wants a simple program to keep track of its employees - regular staff, managers, and interns - and store their records in a text file so the data survives even after the program closes.

This assignment combines three things you've learned separately:
1. **OOP Inheritance** - `Manager` and `Intern` are kinds of `Employee`, so they should reuse what `Employee` already knows how to do instead of repeating code.
2. **File Handling (without `with`)** - you will open files with `open()` and you are responsible for calling `.close()` yourself. There is no safety net here, which is exactly why the next point matters.
3. **Error & Exception Handling** - salaries can't be negative, files might not exist yet, and user input might not even be a number. Your code needs to survive all of that without crashing.

> Note: We are **not** using encapsulation (no leading-underscore "private" attributes) and **not** using polymorphism (child classes will not redefine a parent method with different behaviour). Inheritance here is purely about **reusing** attributes and methods, and **adding new ones** on top.
