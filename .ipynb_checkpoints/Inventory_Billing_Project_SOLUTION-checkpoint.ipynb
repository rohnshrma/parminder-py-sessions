You will build a complete shop management program in **10 steps**. Each step adds one piece. By the end you will have a working program that manages products, customers, a cart, bills and sales reports, with all data saved in `.txt` files.

**Rules for this project**

1. Use only `.txt` files.
2. Open files with `open()` and always close them with `close()`. Do **not** use the `with` statement.
3. Keep all attributes public (for example `self.price`). No private attributes and no `@property`.
4. Do not use any external libraries.
5. Do the steps in order. Every step depends on the ones before it.
6. Run the cells from top to bottom each time you reopen the notebook, so that your classes are defined again.

**How each step is organised**

- **Hints**: small pointers, not solutions

## Business rules (refer back to these)

**Product types and GST**

| Type | GST | Extra detail stored |
|---|---|---|
| `general` | 12% | none (store `-`) |
| `perishable` | 5% | days left before expiry |
| `electronic` | 18% | warranty in months |

**Other rules**

- A product is **low on stock** when its stock is below 5.
- A perishable product with **2 or fewer days left** gets 20% off its price (before GST).
- An electronic product has no special discount.
- Bill discount: bill total of 1000 or more gets 5% off, 5000 or more gets 10% off.
- A **premium** customer gets an extra 5% off on top of the bill discount.

**File formats** (one record per line, values separated by commas)

`products.txt`
```
101,Milk,45.0,20,perishable,3
102,Rice 5kg,420.0,15,general,-
103,Earphones,799.0,8,electronic,6
```
Order: product id, name, price, stock, type, extra detail

`customers.txt`
```
C01,Amit,regular
C02,Neha,premium
```
Order: customer id, name, type

`sales.txt` (one line for each item sold)
```
1001,C02,101,Milk,perishable,2,94.5
```
Order: bill number, customer id, product id, product name, type, quantity, line total

`errors.txt` (one line for each error caught)
```
ProductNotFoundError: No product with id 999
```

## Step 1: Plan the data with basic data types

**Goal:** Before writing any classes, represent the shop's data using plain Python data types, and get comfortable looping over it.

**Tasks**

1. Create these constants: `LOW_STOCK_LIMIT` (5), and a dictionary `GST_RATES` that maps each product type (`"general"`, `"perishable"`, `"electronic"`) to its GST percentage.
2. Create a dictionary `sample_products` with at least 6 products. The key is the product id (int). The value is another dictionary with the keys `name`, `price`, `stock`, `type`.
3. Loop over `sample_products` and print each product on one line, neatly formatted.
4. Build a **set** called `categories` containing every product type that appears in `sample_products`. Print it.
5. Build a **list** called `low_stock_names` with the names of products whose stock is below `LOW_STOCK_LIMIT`. Print it.
6. Create a **tuple** called `line_item` that represents one cart entry: (product id, quantity, price per unit). Print the total for that line using the values inside the tuple.
7. Write a loop that prints the price of each product **including GST**, using `GST_RATES` and if/elif/else or a dictionary lookup.

**Expected result**

- A printed product list, one product per line
- A set such as `{'general', 'perishable', 'electronic'}` (order may differ)
- A list of low-stock product names (make sure at least one of your sample products has stock below 5)
- One line total, and a GST-inclusive price for every product

**Hints**

- `for key, value in my_dict.items():` gives you both the key and the value.
- A set ignores duplicates automatically, so you can simply `add()` every type.
- Price with GST = price + price * rate / 100

````python
# Step 1 - write your code here
LOW_STOCK_LIMIT = 5
GST_RATES = {"general": 12, "perishable": 5, "electronic": 18}

sample_products = {
    101: {"name": "Milk", "price": 45.0, "stock": 20, "type": "perishable"},
    102: {"name": "Rice 5kg", "price": 420.0, "stock": 15, "type": "general"},
    103: {"name": "Earphones", "price": 799.0, "stock": 8, "type": "electronic"},
    104: {"name": "Bread", "price": 35.0, "stock": 3, "type": "perishable"},
    105: {"name": "Notebook", "price": 60.0, "stock": 2, "type": "general"},
    106: {"name": "Charger", "price": 499.0, "stock": 10, "type": "electronic"},
}
````

````python
# Step 1 - test your code here
print("-- All products --")
for product_id, info in sample_products.items():
    print(f"{product_id}: {info['name']} | Rs.{info['price']} | stock {info['stock']} | {info['type']}")

categories = set()
for info in sample_products.values():
    categories.add(info["type"])
print("\n-- Categories --")
print(categories)

low_stock_names = []
for info in sample_products.values():
    if info["stock"] < LOW_STOCK_LIMIT:
        low_stock_names.append(info["name"])
print("\n-- Low stock names --")
print(low_stock_names)

line_item = (101, 3, 45.0)
product_id, qty, price = line_item
print("\n-- Line item total --")
print(qty * price)

print("\n-- Prices including GST --")
for product_id, info in sample_products.items():
    rate = GST_RATES[info["type"]]
    price_with_gst = round(info["price"] + info["price"] * rate / 100, 2)
    print(f"{info['name']}: {price_with_gst}")
````

## Step 2: The `Product` class

**Goal:** Replace the inner dictionaries from Step 1 with a proper class.

**Tasks**

1. Create a class `Product` with these attributes set in `__init__`: `product_id`, `name`, `price`, `stock`. Also set `self.product_type = "general"`.
2. Add a method `gst_rate()` that returns 12.
3. Add a method `final_price()` that returns the price including GST. It must call `self.gst_rate()` rather than using the number 12 directly (this matters in Step 3).
4. Add a method `is_low_stock()` that returns `True` or `False`.
5. Add a method `add_stock(qty)` that increases the stock.
6. Add a method `reduce_stock(qty)` that decreases the stock. For now, if `qty` is more than the stock, just print a message and do not change anything. (You will improve this in Step 4.)
7. Add a method `to_line()` that returns the product as one comma-separated string in the `products.txt` format, for example `102,Rice 5kg,420.0,15,general,-`.
8. Add `__str__` so that `print(product)` shows a readable line with id, name, price, stock and type.
9. Create 3 `Product` objects, store them in a list, and loop over the list to print each one, its final price, and whether it is low on stock.

**Expected result**

- `print(p)` shows a clean, readable line, not something like `<__main__.Product object at 0x...>`
- `p.final_price()` for a price of 420.0 gives 470.4
- `p.to_line()` returns a string with exactly 6 comma-separated values

**Hints**

- `__str__` must **return** a string, not print it.
- Convert numbers with `str()` before joining them with commas.

````python
# Step 2 - write your code here
# NOTE: reduce_stock/add_stock already raise the custom exceptions that
# Step 4 asks for. In a real run you would come back and edit this same
# cell once Step 4 introduces those exceptions, so the solution just
# writes the final version here directly.
class Product:
    def __init__(self, product_id, name, price, stock):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.stock = stock
        self.product_type = "general"

    def gst_rate(self):
        return 12

    def final_price(self):
        return round(self.price + self.price * self.gst_rate() / 100, 2)

    def is_low_stock(self):
        return self.stock < LOW_STOCK_LIMIT

    def add_stock(self, qty):
        if qty <= 0:
            raise InvalidQuantityError("Quantity to add must be greater than 0")
        self.stock += qty

    def reduce_stock(self, qty):
        if qty <= 0:
            raise InvalidQuantityError("Quantity to reduce must be greater than 0")
        if qty > self.stock:
            raise OutOfStockError(f"Not enough stock for {self.name}. Available: {self.stock}")
        self.stock -= qty

    def to_line(self):
        return f"{self.product_id},{self.name},{self.price},{self.stock},{self.product_type},-"

    def __str__(self):
        return f"[{self.product_id}] {self.name} - Rs.{self.price} | Stock: {self.stock} | Type: {self.product_type}"
````

````python
# Step 2 - test your code here
p1 = Product(101, "Milk", 45.0, 20)
p2 = Product(102, "Rice 5kg", 420.0, 15)
p3 = Product(104, "Bread", 35.0, 3)

products_step2 = [p1, p2, p3]
for p in products_step2:
    print(p, "| final price:", p.final_price(), "| low stock:", p.is_low_stock())
````

## Step 3: Inheritance: `PerishableProduct` and `ElectronicProduct`

**Goal:** Create two special kinds of product that reuse `Product` and change only what is different.

**Tasks**

1. Create `PerishableProduct(Product)`.
   - `__init__` takes one extra value, `days_left`. Call `super().__init__(...)` for the common attributes.
   - Set `self.product_type = "perishable"`.
   - Override `gst_rate()` to return 5.
   - Override `final_price()`: if `days_left` is 2 or less, take 20% off the price first, then add GST. Otherwise just add GST.
   - Add a method `is_expired()` that returns `True` when `days_left` is 0 or less.
   - Override `to_line()` so the last value is `days_left` instead of `-`.
   - Override `__str__` to also show the days left. Reuse the parent's text with `super().__str__()`.
2. Create `ElectronicProduct(Product)`.
   - Extra value: `warranty_months`.
   - Set `self.product_type = "electronic"`.
   - Override `gst_rate()` to return 18.
   - Override `to_line()` and `__str__` in the same way.
3. Create a list containing at least one object of **each** of the three classes. Loop over the list and print each product and its `final_price()`.

**Expected result**

- The same loop and the same method call (`final_price()`) give different calculations depending on the object's class. This is polymorphism.
- Milk at 45.0 with 3 days left gives 47.25. The same milk with 1 day left gives 37.8.
- Earphones at 799.0 give 942.82.

**Hints**

- If `final_price()` in `Product` calls `self.gst_rate()`, then `ElectronicProduct` does not need its own `final_price()` at all. Think about why.
- `round(value, 2)` keeps prices to 2 decimal places.

````python
# Step 3 - write your code here
class PerishableProduct(Product):
    def __init__(self, product_id, name, price, stock, days_left):
        super().__init__(product_id, name, price, stock)
        self.days_left = days_left
        self.product_type = "perishable"

    def gst_rate(self):
        return 5

    def final_price(self):
        price = self.price
        if self.days_left <= 2:
            price = price - price * 20 / 100
        return round(price + price * self.gst_rate() / 100, 2)

    def is_expired(self):
        return self.days_left <= 0

    def to_line(self):
        return f"{self.product_id},{self.name},{self.price},{self.stock},{self.product_type},{self.days_left}"

    def __str__(self):
        return super().__str__() + f" | Days left: {self.days_left}"


class ElectronicProduct(Product):
    def __init__(self, product_id, name, price, stock, warranty_months):
        super().__init__(product_id, name, price, stock)
        self.warranty_months = warranty_months
        self.product_type = "electronic"

    def gst_rate(self):
        return 18

    def to_line(self):
        return f"{self.product_id},{self.name},{self.price},{self.stock},{self.product_type},{self.warranty_months}"

    def __str__(self):
        return super().__str__() + f" | Warranty: {self.warranty_months} months"
````

````python
# Step 3 - test your code here
milk_fresh = PerishableProduct(101, "Milk", 45.0, 20, 3)
milk_old = PerishableProduct(107, "Milk (old)", 45.0, 20, 1)
earphones = ElectronicProduct(103, "Earphones", 799.0, 8, 6)
rice = Product(102, "Rice 5kg", 420.0, 15)

mixed_products = [rice, milk_fresh, milk_old, earphones]
for p in mixed_products:
    print(p, "| final price:", p.final_price())
````

## Step 4: Custom exceptions and safe input

**Goal:** Make the program report problems with exceptions instead of print messages, and make sure wrong user input can never crash it.

**Tasks**

1. Create three custom exception classes that inherit from `Exception`:
   - `ProductNotFoundError`
   - `OutOfStockError`
   - `InvalidQuantityError`
2. Go back to your `Product` class in Step 2 and update `reduce_stock(qty)`:
   - If `qty` is 0 or negative, raise `InvalidQuantityError` with a clear message.
   - If `qty` is more than the stock, raise `OutOfStockError` with a message that includes the product name and the available stock.
   - Otherwise reduce the stock.
   - Update `add_stock(qty)` to raise `InvalidQuantityError` for 0 or negative values too.
   - Re-run the Step 2 and Step 3 cells afterwards so the child classes pick up the change.
3. Write a function `get_int(message)` that keeps asking the user until they type a valid whole number, then returns it. Use `while True` and `try/except ValueError`.
4. Write a function `get_float(message)` that does the same for decimal numbers.
5. Write a function `get_positive_int(message)` that uses `get_int` and keeps asking until the number is greater than 0.
6. Write a function `get_choice(message, valid_options)` that keeps asking until the user's answer is in the list `valid_options`.
7. Test `reduce_stock` three times inside `try/except/else/finally`: once with a valid quantity, once with too large a quantity, once with a negative quantity. In `else`, print a success message. In `finally`, print the current stock.

**Expected result**

- Typing `abc` when a number is expected shows a friendly message and asks again. The program does not stop.
- Each wrong `reduce_stock` call prints the message from your custom exception, and the stock stays unchanged.
- The `finally` line prints every time, whether the call worked or failed.

**Hints**

- A custom exception can be as short as a class with `pass` inside.
- `except OutOfStockError as e:` lets you print the message with `print(e)`.
- You can catch two exception types in one line: `except (TypeA, TypeB) as e:`

````python
# Step 4 - write your code here
class ProductNotFoundError(Exception):
    pass


class OutOfStockError(Exception):
    pass


class InvalidQuantityError(Exception):
    pass


def get_int(message):
    while True:
        value = input(message)
        try:
            return int(value)
        except ValueError:
            print("Please enter a whole number.")


def get_float(message):
    while True:
        value = input(message)
        try:
            return float(value)
        except ValueError:
            print("Please enter a valid number.")


def get_positive_int(message):
    while True:
        value = get_int(message)
        if value > 0:
            return value
        print("Please enter a number greater than 0.")


def get_choice(message, valid_options):
    while True:
        value = input(message)
        if value in valid_options:
            return value
        print(f"Please choose one of: {', '.join(valid_options)}")
````

````python
# Step 4 - test your code here
test_product = Product(201, "Test Item", 100.0, 10)

try:
    test_product.reduce_stock(4)
except (OutOfStockError, InvalidQuantityError) as e:
    print("Error:", e)
else:
    print("Reduced stock successfully.")
finally:
    print("Stock after call 1:", test_product.stock)

try:
    test_product.reduce_stock(100)
except (OutOfStockError, InvalidQuantityError) as e:
    print("Error:", e)
else:
    print("Reduced stock successfully.")
finally:
    print("Stock after call 2:", test_product.stock)

try:
    test_product.reduce_stock(-5)
except (OutOfStockError, InvalidQuantityError) as e:
    print("Error:", e)
else:
    print("Reduced stock successfully.")
finally:
    print("Stock after call 3:", test_product.stock)

# get_int/get_float/get_positive_int/get_choice read real keyboard input,
# so they are left commented out in this automated solution. Uncomment
# one at a time in Jupyter to try them interactively:
# age = get_int("Enter your age: ")
# price = get_float("Enter a price: ")
# qty = get_positive_int("Enter a quantity: ")
# answer = get_choice("yes or no? ", ["yes", "no"])
````

## Step 5: The `Inventory` class

**Goal:** Build one object that holds and manages all products.

**Tasks**

1. Create a class `Inventory`. In `__init__`, create an empty dictionary `self.products` (key: product id, value: a product object).
2. Add these methods:
   - `add_product(product)`: add the object to the dictionary. If the id already exists, raise `ValueError` with a clear message.
   - `get_product(product_id)`: return the product. If the id does not exist, raise `ProductNotFoundError`.
   - `delete_product(product_id)`: remove it. Use `get_product` first so a wrong id raises the error.
   - `update_price(product_id, new_price)`: raise `ValueError` if the new price is 0 or negative.
   - `restock(product_id, qty)`: use the product's `add_stock`.
   - `view_all()`: print every product. If there are none, print a message saying the inventory is empty.
   - `search_by_name(text)`: return a **list** of products whose name contains `text`. The search should ignore upper and lower case.
   - `low_stock_products()`: return a list of products that are low on stock.
   - `categories()`: return a **set** of all product types in the inventory.
   - `products_by_type(product_type)`: return a list of products of that type.
3. Create an `Inventory`, add at least 6 products of mixed types, and test every method. Wrap the calls that can fail in `try/except`.

**Expected result**

- Adding the same id twice shows your `ValueError` message.
- Getting or deleting an id that does not exist shows your `ProductNotFoundError` message.
- Searching `"ri"` finds `Rice 5kg`.
- `categories()` returns a set with no repeated values.

**Hints**

- `if product_id in self.products:` checks whether a key exists.
- `del self.products[product_id]` removes a key.
- `.lower()` on both strings makes a search ignore case.

````python
# Step 5 - write your code here
class Inventory:
    def __init__(self):
        self.products = {}

    def add_product(self, product):
        if product.product_id in self.products:
            raise ValueError(f"Product id {product.product_id} already exists")
        self.products[product.product_id] = product

    def get_product(self, product_id):
        if product_id not in self.products:
            raise ProductNotFoundError(f"No product with id {product_id}")
        return self.products[product_id]

    def delete_product(self, product_id):
        product = self.get_product(product_id)
        del self.products[product.product_id]

    def update_price(self, product_id, new_price):
        if new_price <= 0:
            raise ValueError("Price must be greater than 0")
        product = self.get_product(product_id)
        product.price = new_price

    def restock(self, product_id, qty):
        product = self.get_product(product_id)
        product.add_stock(qty)

    def view_all(self):
        if not self.products:
            print("Inventory is empty.")
            return
        for product in self.products.values():
            print(product)

    def search_by_name(self, text):
        result = []
        for product in self.products.values():
            if text.lower() in product.name.lower():
                result.append(product)
        return result

    def low_stock_products(self):
        result = []
        for product in self.products.values():
            if product.is_low_stock():
                result.append(product)
        return result

    def categories(self):
        result = set()
        for product in self.products.values():
            result.add(product.product_type)
        return result

    def products_by_type(self, product_type):
        result = []
        for product in self.products.values():
            if product.product_type == product_type:
                result.append(product)
        return result

    # save_to_file/load_from_file call log_error, which Step 6 defines
    # below -- only call these two methods after that cell has run.
    def save_to_file(self, filename):
        f = None
        try:
            f = open(filename, "w")
            for product in self.products.values():
                f.write(product.to_line() + "\n")
        finally:
            if f is not None:
                f.close()

    def load_from_file(self, filename):
        f = None
        try:
            f = open(filename, "r")
        except FileNotFoundError:
            print(f"{filename} not found. Starting with an empty inventory.")
            return
        try:
            line_number = 0
            for line in f.readlines():
                line_number += 1
                line = line.strip()
                if line == "":
                    continue
                try:
                    parts = line.split(",")
                    if len(parts) != 6:
                        raise ValueError("Wrong number of values")
                    product_id = int(parts[0])
                    name = parts[1]
                    price = float(parts[2])
                    stock = int(parts[3])
                    product_type = parts[4]
                    extra = parts[5]
                    if product_type == "perishable":
                        product = PerishableProduct(product_id, name, price, stock, int(extra))
                    elif product_type == "electronic":
                        product = ElectronicProduct(product_id, name, price, stock, int(extra))
                    else:
                        product = Product(product_id, name, price, stock)
                    self.products[product_id] = product
                except (ValueError, IndexError) as e:
                    log_error(f"Line {line_number} in {filename}: {e}")
        finally:
            f.close()
````

````python
# Step 5 - test your code here
inventory = Inventory()
inventory.add_product(Product(102, "Rice 5kg", 420.0, 15))
inventory.add_product(PerishableProduct(101, "Milk", 45.0, 20, 3))
inventory.add_product(ElectronicProduct(103, "Earphones", 799.0, 8, 6))
inventory.add_product(PerishableProduct(104, "Bread", 35.0, 3, 1))
inventory.add_product(Product(105, "Notebook", 60.0, 2))
inventory.add_product(ElectronicProduct(106, "Charger", 499.0, 10, 12))

print("-- All products --")
inventory.view_all()

try:
    inventory.add_product(Product(102, "Duplicate Rice", 100.0, 5))
except ValueError as e:
    print("\nExpected error:", e)

try:
    inventory.get_product(999)
except ProductNotFoundError as e:
    print("Expected error:", e)

try:
    inventory.delete_product(999)
except ProductNotFoundError as e:
    print("Expected error:", e)

print("\n-- Search 'ri' --")
for p in inventory.search_by_name("ri"):
    print(p)

print("\n-- Low stock --")
for p in inventory.low_stock_products():
    print(p)

print("\n-- Categories --")
print(inventory.categories())

print("\n-- Electronics only --")
for p in inventory.products_by_type("electronic"):
    print(p)
````

   - Loop over all products and write each product's `to_line()` followed by a new line.

3. Add a method `load_from_file(filename)` to `Inventory`:
   - Open the file in read mode. If it does not exist, catch `FileNotFoundError`, print a message that you are starting with an empty inventory, and return.

4. Test it:

**Expected result**

- `products.txt` has one line per product, 6 values per line.
- The loaded inventory prints exactly the same products as the original, with the correct class for each (a perishable product still shows its days left).
- A damaged line is skipped and recorded in `errors.txt`.

**Hints**

- Set `f = None` before the `try`, then in `finally` write `if f is not None: f.close()`. Think about why this check is needed when the file does not exist.
- Splitting `"101,Milk,45.0,20,perishable,3"` by comma gives a list of 6 strings. Everything read from a file is a string until you convert it.
- `len(parts) != 6` is a quick test for a damaged line. You can `raise ValueError` yourself in that case and let the same `except` handle it.
- Wrong index and wrong conversion raise different errors: `IndexError` and `ValueError`.

````python
# Step 6 - write your code here
def log_error(message):
    f = None
    try:
        f = open("errors.txt", "a")
        f.write(message + "\n")
    finally:
        if f is not None:
            f.close()
````

## Step 7: Customers and `customers.txt`

**Goal:** Add customers, with a premium type that gets an extra discount, and store them in a file.

**Tasks**

1. Create a class `Customer` with attributes `customer_id`, `name`, and `customer_type = "regular"`.
   - Method `extra_discount()` returns 0.
   - Method `to_line()` returns a line in the `customers.txt` format.
   - `__str__` shows id, name and type.
2. Create `PremiumCustomer(Customer)`.
   - Set `customer_type = "premium"`.
   - Override `extra_discount()` to return 5.
3. Create a custom exception `CustomerNotFoundError`.
4. Create a class `CustomerBook` with a dictionary `self.customers` and these methods:
   - `add_customer(customer)`: raise `ValueError` if the id already exists.
   - `get_customer(customer_id)`: raise `CustomerNotFoundError` if it is missing.
   - `view_all()`
   - `save_to_file(filename)` and `load_from_file(filename)`, following the same pattern as Step 6 (create the right class based on the type, skip and log damaged lines, close in `finally`).
5. Test it: add 3 customers (at least one premium), save, load into a new `CustomerBook`, and print them. Loop over all customers and print each one's extra discount.

**Expected result**

- `customers.txt` has one line per customer with 3 values.
- After loading, premium customers are still `PremiumCustomer` objects and return 5 from `extra_discount()`.
- Asking for an id that does not exist shows your `CustomerNotFoundError` message.

**Hints**

- This step is deliberately similar to Steps 5 and 6. Try to write it without looking back, then compare.

````python
# Step 7 - write your code here
class CustomerNotFoundError(Exception):
    pass


class Customer:
    def __init__(self, customer_id, name):
        self.customer_id = customer_id
        self.name = name
        self.customer_type = "regular"

    def extra_discount(self):
        return 0

    def to_line(self):
        return f"{self.customer_id},{self.name},{self.customer_type}"

    def __str__(self):
        return f"[{self.customer_id}] {self.name} ({self.customer_type})"


class PremiumCustomer(Customer):
    def __init__(self, customer_id, name):
        super().__init__(customer_id, name)
        self.customer_type = "premium"

    def extra_discount(self):
        return 5


class CustomerBook:
    def __init__(self):
        self.customers = {}

    def add_customer(self, customer):
        if customer.customer_id in self.customers:
            raise ValueError(f"Customer id {customer.customer_id} already exists")
        self.customers[customer.customer_id] = customer

    def get_customer(self, customer_id):
        if customer_id not in self.customers:
            raise CustomerNotFoundError(f"No customer with id {customer_id}")
        return self.customers[customer_id]

    def view_all(self):
        if not self.customers:
            print("No customers yet.")
            return
        for customer in self.customers.values():
            print(customer)

    def save_to_file(self, filename):
        f = None
        try:
            f = open(filename, "w")
            for customer in self.customers.values():
                f.write(customer.to_line() + "\n")
        finally:
            if f is not None:
                f.close()

    def load_from_file(self, filename):
        f = None
        try:
            f = open(filename, "r")
        except FileNotFoundError:
            print(f"{filename} not found. Starting with no customers.")
            return
        try:
            line_number = 0
            for line in f.readlines():
                line_number += 1
                line = line.strip()
                if line == "":
                    continue
                try:
                    parts = line.split(",")
                    if len(parts) != 3:
                        raise ValueError("Wrong number of values")
                    customer_id, name, customer_type = parts
                    if customer_type == "premium":
                        customer = PremiumCustomer(customer_id, name)
                    else:
                        customer = Customer(customer_id, name)
                    self.customers[customer_id] = customer
                except (ValueError, IndexError) as e:
                    log_error(f"Line {line_number} in {filename}: {e}")
        finally:
            f.close()
````

````python
# Step 7 - test your code here
customer_book = CustomerBook()
customer_book.add_customer(Customer("C01", "Amit"))
customer_book.add_customer(PremiumCustomer("C02", "Neha"))
customer_book.add_customer(Customer("C03", "Priya"))

customer_book.save_to_file("customers.txt")

reloaded_customers = CustomerBook()
reloaded_customers.load_from_file("customers.txt")
print("-- Reloaded customers --")
reloaded_customers.view_all()

print("\n-- Extra discount for each customer --")
for customer in reloaded_customers.customers.values():
    print(customer.name, "->", customer.extra_discount())

try:
    reloaded_customers.get_customer("C99")
except CustomerNotFoundError as e:
    print("\nExpected error:", e)
````

## Step 8: The `Cart` class

**Goal:** Let a customer collect items before paying, with full stock checking.

**Tasks**

1. Create a class `Cart`. `__init__` takes a `customer` object and an `inventory` object and creates an empty list `self.items`. Each entry in the list is a **tuple**: (product object, quantity).
2. Add these methods:
   - `add_item(product_id, qty)`:
     - Get the product from the inventory (this may raise `ProductNotFoundError`).
     - Raise `InvalidQuantityError` if `qty` is 0 or negative.
     - If the product is a perishable product and is expired, raise `ValueError`.
     - Work out how many of this product are **already in the cart**. If already-in-cart plus `qty` is more than the stock, raise `OutOfStockError`.
     - If the product is already in the cart, replace its tuple with a new one holding the increased quantity. Otherwise append a new tuple.
     - Do **not** reduce the stock here. Stock changes only at checkout.
   - `remove_item(product_id)`: remove that product from the cart. Raise `ProductNotFoundError` if it is not in the cart.
   - `is_empty()`: return `True` or `False`.
   - `subtotal()`: return the sum of `final_price() * qty` for all items.
   - `discount_percent()`: return 0, 5 or 10 based on the subtotal (see the business rules), plus the customer's `extra_discount()`.
   - `total()`: return the subtotal after the discount.
   - `view()`: print every item with its quantity and line total, then the subtotal, discount percent and final total. Print a message if the cart is empty.
   - `clear()`: empty the cart.
3. Test it with a regular customer and a premium customer. Try adding a wrong id, a negative quantity, and more than the stock. Add the same product twice and check that it appears once with the combined quantity.

**Expected result**

- Every wrong action shows the right custom error message, and the cart stays unchanged.
- The same cart contents give a lower total for a premium customer.
- A cart with a subtotal of 5200 for a premium customer shows a 15% discount.

**Hints**

- Tuples cannot be changed. To update a quantity, build a new tuple and put it at the same position in the list.
- `for i in range(len(self.items)):` gives you the position, which you need for replacing or removing.
- `isinstance(product, PerishableProduct)` tells you whether an object belongs to a class.

````python
# Step 8 - write your code here
class Cart:
    def __init__(self, customer, inventory):
        self.customer = customer
        self.inventory = inventory
        self.items = []

    def _find_index(self, product_id):
        for i in range(len(self.items)):
            if self.items[i][0].product_id == product_id:
                return i
        return -1

    def add_item(self, product_id, qty):
        product = self.inventory.get_product(product_id)
        if qty <= 0:
            raise InvalidQuantityError("Quantity must be greater than 0")
        if isinstance(product, PerishableProduct) and product.is_expired():
            raise ValueError(f"{product.name} has expired and cannot be sold")
        index = self._find_index(product_id)
        already_in_cart = 0
        if index != -1:
            already_in_cart = self.items[index][1]
        if already_in_cart + qty > product.stock:
            raise OutOfStockError(f"Not enough stock for {product.name}. Available: {product.stock}")
        if index != -1:
            self.items[index] = (product, already_in_cart + qty)
        else:
            self.items.append((product, qty))

    def remove_item(self, product_id):
        index = self._find_index(product_id)
        if index == -1:
            raise ProductNotFoundError(f"Product {product_id} is not in the cart")
        del self.items[index]

    def is_empty(self):
        return len(self.items) == 0

    def subtotal(self):
        total = 0
        for product, qty in self.items:
            total += product.final_price() * qty
        return round(total, 2)

    def discount_percent(self):
        subtotal = self.subtotal()
        if subtotal >= 5000:
            percent = 10
        elif subtotal >= 1000:
            percent = 5
        else:
            percent = 0
        return percent + self.customer.extra_discount()

    def total(self):
        subtotal = self.subtotal()
        discount = self.discount_percent()
        return round(subtotal - subtotal * discount / 100, 2)

    def view(self):
        if self.is_empty():
            print("Cart is empty.")
            return
        for product, qty in self.items:
            line_total = round(product.final_price() * qty, 2)
            print(f"{product.name:<20}{qty:>5}{line_total:>12.2f}")
        print("-" * 40)
        print(f"Subtotal: {self.subtotal():.2f}")
        print(f"Discount: {self.discount_percent()}%")
        print(f"Total: {self.total():.2f}")

    def clear(self):
        self.items = []
````

## Step 9: Checkout: bill file and sales record

**Goal:** Complete a sale. Reduce the stock, write a bill to its own text file, and record the sale in `sales.txt`.

**Tasks**

1. Write a function `next_bill_number(filename)`:
   - Read `sales.txt` and find the largest bill number used so far. Return that number plus 1.
   - If the file does not exist or is empty, return 1001.
2. Write a function `checkout(cart, inventory)` that does the following in order:
   - If the cart is empty, raise `ValueError`.
   - Get the next bill number.
   - Loop over the cart and call `reduce_stock` on each product.
   - Create a file named `bill_<number>.txt` (for example `bill_1001.txt`) and write a neat bill into it: shop name, bill number, customer name and type, a header row, one row per item (name, quantity, unit price with GST, line total), then subtotal, discount percent, discount amount and final total.
   - Append one line per item to `sales.txt` in the format given in the business rules.
   - Save the inventory to `products.txt` so the new stock is permanent.
   - Clear the cart.
   - Return the bill number.
   - Every file must be closed in `finally`. Any error must be passed to `log_error`.
3. Write a function `show_bill(bill_number)` that opens the bill file, prints its content with `read()`, and handles the case where the bill does not exist.
4. Test it: build a cart with 3 items, check out, show the bill, then view the inventory to confirm the stock went down. Check out a second cart and confirm that the bill number increased by 1.

**Expected result**

- A new file `bill_1001.txt` that looks like a real receipt, with columns lined up.
- `sales.txt` grows by one line per item sold, and old lines are never lost.
- The stock in `products.txt` matches what is left after the sale.
- Checking out an empty cart shows an error message and creates no files.

**Hints**

- Take the subtotal, discount and total **before** you clear the cart.
- f-strings can align columns: `f"{name:<15}{qty:>5}{price:>10.2f}"`
- `"-" * 40` makes a separator line.
- Append mode `"a"` creates the file if it does not exist, and keeps old content if it does.

````python
# Step 9 - write your code here
def next_bill_number(filename):
    f = None
    try:
        f = open(filename, "r")
    except FileNotFoundError:
        return 1001
    try:
        max_number = 0
        for line in f.readlines():
            line = line.strip()
            if line == "":
                continue
            parts = line.split(",")
            try:
                bill_number = int(parts[0])
                if bill_number > max_number:
                    max_number = bill_number
            except (ValueError, IndexError):
                continue
        if max_number == 0:
            return 1001
        return max_number + 1
    finally:
        f.close()


def checkout(cart, inventory):
    if cart.is_empty():
        raise ValueError("Cannot checkout an empty cart")

    bill_number = next_bill_number("sales.txt")
    subtotal = cart.subtotal()
    discount_percent = cart.discount_percent()
    discount_amount = round(subtotal * discount_percent / 100, 2)
    total = cart.total()
    items_snapshot = list(cart.items)

    for product, qty in items_snapshot:
        product.reduce_stock(qty)

    bill_file = None
    try:
        bill_file = open(f"bill_{bill_number}.txt", "w")
        bill_file.write("===== WebiGeeks Store =====\n")
        bill_file.write(f"Bill No: {bill_number}\n")
        bill_file.write(f"Customer: {cart.customer.name} ({cart.customer.customer_type})\n")
        bill_file.write("-" * 40 + "\n")
        bill_file.write(f"{'Item':<20}{'Qty':>5}{'Price':>8}{'Total':>10}\n")
        bill_file.write("-" * 40 + "\n")
        for product, qty in items_snapshot:
            unit_price = product.final_price()
            line_total = round(unit_price * qty, 2)
            bill_file.write(f"{product.name:<20}{qty:>5}{unit_price:>8.2f}{line_total:>10.2f}\n")
        bill_file.write("-" * 40 + "\n")
        bill_file.write(f"Subtotal: {subtotal:.2f}\n")
        bill_file.write(f"Discount: {discount_percent}% (-{discount_amount:.2f})\n")
        bill_file.write(f"Total: {total:.2f}\n")
    finally:
        if bill_file is not None:
            bill_file.close()

    sales_file = None
    try:
        sales_file = open("sales.txt", "a")
        for product, qty in items_snapshot:
            line_total = round(product.final_price() * qty, 2)
            sales_file.write(
                f"{bill_number},{cart.customer.customer_id},{product.product_id},"
                f"{product.name},{product.product_type},{qty},{line_total}\n"
            )
    finally:
        if sales_file is not None:
            sales_file.close()

    inventory.save_to_file("products.txt")
    cart.clear()
    return bill_number


def show_bill(bill_number):
    f = None
    try:
        f = open(f"bill_{bill_number}.txt", "r")
        print(f.read())
    except FileNotFoundError:
        print(f"Bill {bill_number} does not exist.")
    finally:
        if f is not None:
            f.close()
````

````python
# Step 9 - test your code here
checkout_cart = Cart(premium_customer, reloaded_inventory)
checkout_cart.add_item(101, 2)
checkout_cart.add_item(102, 1)
checkout_cart.add_item(103, 1)

bill_number = checkout(checkout_cart, reloaded_inventory)
print("Bill number:", bill_number)
show_bill(bill_number)

print("\n-- Inventory after checkout --")
reloaded_inventory.view_all()

second_cart = Cart(regular_customer, reloaded_inventory)
second_cart.add_item(104, 1)
second_bill_number = checkout(second_cart, reloaded_inventory)
print("\nSecond bill number:", second_bill_number)
assert second_bill_number == bill_number + 1
print("Bill numbers increase correctly.")

try:
    checkout(Cart(regular_customer, reloaded_inventory), reloaded_inventory)
except ValueError as e:
    print("\nExpected error for empty cart:", e)
````

## Step 10: Reports and the main menu (`Store` class)

**Goal:** Read the sales history to produce reports, then connect everything into one menu-driven program.

**Part A: Reports**

Write a function `load_sales(filename)` that reads `sales.txt` and returns a **list of tuples**, one per line, with every value converted to the right type. Handle a missing file (return an empty list) and skip damaged lines.

Then write these report functions. Each one takes the list of sales.

1. `total_revenue(sales)`: print the total amount sold and the number of **different** bills (use a set).
2. `category_report(sales)`: for each category, print the total quantity and total amount. Use a **nested loop**: the outer loop goes over a set of categories, the inner loop goes over the sales.
3. `top_products(sales, n)`: print the `n` products with the highest total quantity sold. First build a dictionary of product name to total quantity. Then find the top `n` using loops only.
4. `customer_history(sales, customer_id)`: print every item that customer bought, and their total spending. Print a message if they have no purchases.

**Part B: The `Store` class and menu**

1. Create a class `Store`. In `__init__`, create an `Inventory` and a `CustomerBook`, and load both from their files.

```

1. Product management
```

3. Each option opens its own sub-menu (a separate method for each keeps the code readable):

   - **Customer management:** add customer, view all, back
   - **New sale:** ask for the customer id, create a `Cart`, then loop: add item, remove item, view cart, checkout, cancel
   - **Reports:** the four reports from Part A, back
4. Rules for the menu:

   - Every action that can fail must be inside `try/except`. Show the error message, call `log_error`, and return to the menu. The program must never crash.
   - Save to file after every change to products or customers.
   - Option 6 saves everything, prints a goodbye message, and uses `break`.
5. Create a `Store` object and call `run()`.

**Final testing checklist**

Go through this list. Every line should work without a crash.

**Hints**

- To find the top `n` without sorting: repeat `n` times, each time loop over the dictionary to find the largest value, print it, then remove that key.

````python
# Step 10 - write your code here
def load_sales(filename):
    f = None
    try:
        f = open(filename, "r")
    except FileNotFoundError:
        return []
    try:
        sales = []
        for line in f.readlines():
            line = line.strip()
            if line == "":
                continue
            try:
                parts = line.split(",")
                if len(parts) != 7:
                    raise ValueError("Wrong number of values")
                bill_number = int(parts[0])
                customer_id = parts[1]
                product_id = int(parts[2])
                product_name = parts[3]
                product_type = parts[4]
                qty = int(parts[5])
                line_total = float(parts[6])
                sales.append((bill_number, customer_id, product_id, product_name, product_type, qty, line_total))
            except (ValueError, IndexError) as e:
                log_error(f"Bad sales line: {line} ({e})")
        return sales
    finally:
        f.close()


def total_revenue(sales):
    total = 0
    bill_numbers = set()
    for sale in sales:
        total += sale[6]
        bill_numbers.add(sale[0])
    print(f"Total revenue: {total:.2f}")
    print(f"Number of bills: {len(bill_numbers)}")


def category_report(sales):
    categories = set()
    for sale in sales:
        categories.add(sale[4])
    for category in categories:
        total_qty = 0
        total_amount = 0
        for sale in sales:
            if sale[4] == category:
                total_qty += sale[5]
                total_amount += sale[6]
        print(f"{category:<12} qty: {total_qty:<6} amount: {total_amount:.2f}")


def top_products(sales, n):
    totals = {}
    for sale in sales:
        name = sale[3]
        qty = sale[5]
        if name in totals:
            totals[name] += qty
        else:
            totals[name] = qty

    remaining = dict(totals)
    count = 0
    while count < n and remaining:
        best_name = None
        best_qty = -1
        for name, qty in remaining.items():
            if qty > best_qty:
                best_qty = qty
                best_name = name
        print(f"{best_name}: {best_qty}")
        del remaining[best_name]
        count += 1


def customer_history(sales, customer_id):
    found = False
    total_spent = 0
    for sale in sales:
        if sale[1] == customer_id:
            found = True
            print(f"{sale[3]} x{sale[5]} = {sale[6]:.2f}")
            total_spent += sale[6]
    if not found:
        print(f"Customer {customer_id} has no purchases.")
    else:
        print(f"Total spent: {total_spent:.2f}")


class Store:
    def __init__(self):
        self.inventory = Inventory()
        self.customers = CustomerBook()
        self.inventory.load_from_file("products.txt")
        self.customers.load_from_file("customers.txt")

    def run(self):
        while True:
            print("\n===== WebiGeeks Store =====")
            print("1. Product management")
            print("2. Customer management")
            print("3. New sale")
            print("4. Reports")
            print("5. View a bill")
            print("6. Exit")
            choice = get_choice("Choose an option: ", ["1", "2", "3", "4", "5", "6"])
            if choice == "1":
                self.product_menu()
            elif choice == "2":
                self.customer_menu()
            elif choice == "3":
                self.sale_menu()
            elif choice == "4":
                self.report_menu()
            elif choice == "5":
                bill_number = get_int("Bill number: ")
                show_bill(bill_number)
            elif choice == "6":
                self.inventory.save_to_file("products.txt")
                self.customers.save_to_file("customers.txt")
                print("Goodbye!")
                break

    def product_menu(self):
        while True:
            print("\n-- Product management --")
            print("1. Add product")
            print("2. View all")
            print("3. Search by name")
            print("4. Update price")
            print("5. Restock")
            print("6. Delete product")
            print("7. Low stock list")
            print("8. Back")
            choice = get_choice("Choose an option: ", ["1", "2", "3", "4", "5", "6", "7", "8"])
            try:
                if choice == "1":
                    product_type = get_choice(
                        "Type (general/perishable/electronic): ",
                        ["general", "perishable", "electronic"],
                    )
                    product_id = get_positive_int("Product id: ")
                    name = input("Name: ")
                    price = get_float("Price: ")
                    stock = get_positive_int("Stock: ")
                    if product_type == "perishable":
                        days_left = get_int("Days left: ")
                        product = PerishableProduct(product_id, name, price, stock, days_left)
                    elif product_type == "electronic":
                        warranty_months = get_int("Warranty (months): ")
                        product = ElectronicProduct(product_id, name, price, stock, warranty_months)
                    else:
                        product = Product(product_id, name, price, stock)
                    self.inventory.add_product(product)
                    self.inventory.save_to_file("products.txt")
                    print("Product added.")
                elif choice == "2":
                    self.inventory.view_all()
                elif choice == "3":
                    text = input("Search text: ")
                    results = self.inventory.search_by_name(text)
                    if not results:
                        print("No products found.")
                    for product in results:
                        print(product)
                elif choice == "4":
                    product_id = get_int("Product id: ")
                    new_price = get_float("New price: ")
                    self.inventory.update_price(product_id, new_price)
                    self.inventory.save_to_file("products.txt")
                    print("Price updated.")
                elif choice == "5":
                    product_id = get_int("Product id: ")
                    qty = get_positive_int("Quantity to add: ")
                    self.inventory.restock(product_id, qty)
                    self.inventory.save_to_file("products.txt")
                    print("Stock updated.")
                elif choice == "6":
                    product_id = get_int("Product id: ")
                    self.inventory.delete_product(product_id)
                    self.inventory.save_to_file("products.txt")
                    print("Product deleted.")
                elif choice == "7":
                    results = self.inventory.low_stock_products()
                    if not results:
                        print("No low-stock products.")
                    for product in results:
                        print(product)
                elif choice == "8":
                    break
            except (ValueError, ProductNotFoundError, InvalidQuantityError, OutOfStockError) as e:
                print(f"Error: {e}")
                log_error(f"{type(e).__name__}: {e}")

    def customer_menu(self):
        while True:
            print("\n-- Customer management --")
            print("1. Add customer")
            print("2. View all")
            print("3. Back")
            choice = get_choice("Choose an option: ", ["1", "2", "3"])
            try:
                if choice == "1":
                    customer_id = input("Customer id: ")
                    name = input("Name: ")
                    customer_type = get_choice("Type (regular/premium): ", ["regular", "premium"])
                    if customer_type == "premium":
                        customer = PremiumCustomer(customer_id, name)
                    else:
                        customer = Customer(customer_id, name)
                    self.customers.add_customer(customer)
                    self.customers.save_to_file("customers.txt")
                    print("Customer added.")
                elif choice == "2":
                    self.customers.view_all()
                elif choice == "3":
                    break
            except (ValueError, CustomerNotFoundError) as e:
                print(f"Error: {e}")
                log_error(f"{type(e).__name__}: {e}")

    def sale_menu(self):
        try:
            customer_id = input("Customer id: ")
            customer = self.customers.get_customer(customer_id)
        except CustomerNotFoundError as e:
            print(f"Error: {e}")
            log_error(f"{type(e).__name__}: {e}")
            return

        cart = Cart(customer, self.inventory)
        while True:
            print("\n-- New sale --")
            print("1. Add item")
            print("2. Remove item")
            print("3. View cart")
            print("4. Checkout")
            print("5. Cancel")
            choice = get_choice("Choose an option: ", ["1", "2", "3", "4", "5"])
            try:
                if choice == "1":
                    product_id = get_int("Product id: ")
                    qty = get_positive_int("Quantity: ")
                    cart.add_item(product_id, qty)
                    print("Added to cart.")
                elif choice == "2":
                    product_id = get_int("Product id: ")
                    cart.remove_item(product_id)
                    print("Removed from cart.")
                elif choice == "3":
                    cart.view()
                elif choice == "4":
                    bill_number = checkout(cart, self.inventory)
                    print(f"Sale complete. Bill number: {bill_number}")
                    break
                elif choice == "5":
                    print("Sale cancelled.")
                    break
            except (ValueError, ProductNotFoundError, InvalidQuantityError, OutOfStockError) as e:
                print(f"Error: {e}")
                log_error(f"{type(e).__name__}: {e}")

    def report_menu(self):
        while True:
            print("\n-- Reports --")
            print("1. Total revenue")
            print("2. Category report")
            print("3. Top products")
            print("4. Customer history")
            print("5. Back")
            choice = get_choice("Choose an option: ", ["1", "2", "3", "4", "5"])
            sales = load_sales("sales.txt")
            if choice == "1":
                total_revenue(sales)
            elif choice == "2":
                category_report(sales)
            elif choice == "3":
                n = get_positive_int("How many top products? ")
                top_products(sales, n)
            elif choice == "4":
                customer_id = input("Customer id: ")
                customer_history(sales, customer_id)
            elif choice == "5":
                break
````
