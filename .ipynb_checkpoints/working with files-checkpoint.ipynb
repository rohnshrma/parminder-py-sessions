```python
def fileManager(path):
    print("==== FILE MANAGER====")
    while True:
        print("\n==== Menu ====")
        print("1. Read File")
        print("2. Write File")
        print("3. Add Into File")
        print("4. Exit")

        op = int(input("\nEnter your choice (1-4) :"))

        if  op < 1 or op > 4:
            print("Invalid Operation")
            continue

        if op == 1:
            file = open(path, "r")
            data = file.read()
            print(f"DATA => \n{data}")
            file.close()
        elif op == 2:
            file = open(path, "w")
            data = input("enter data to overwrite : ")
            file.write(data + "\n")
            print("File OverWritten Successfully")
            file.close()
        elif op == 3:
            file = open(path, "a")
            data = input("enter data to add : ")
            file.write(data + "\n")
            print("File Written Successfully")
            file.close()
        else :
            print("Thanks for using file manager!")
            break
            
```

```python
fileManager("./DATA/student.txt")
```
