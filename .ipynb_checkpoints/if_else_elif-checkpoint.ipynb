# if
the code inside of the if blocks runs only when the given condition is True



```python

if condition:
    # code to be executed

```

````python
# input() always gives back text (a string), so we wrap it in int() to turn it into a whole number
# otherwise we can't compare it with >= like a number
age = int(input("Enter your age : "))

# this is the actual "if" check, python reads it like a question: "is age 18 or more?"
# the colon at the end just means "here comes the block of code that runs if this is true"
if age >= 18:
    # this line is indented (pushed in with spaces), and that indentation is what tells python
    # "this line belongs to the if statement above", it will ONLY run when the condition is True
    print("Eligible to drive")

# note: since we typed 12 above, 12 >= 18 is False, so the print line never ran, nothing got printed
````

# else
the code inside of the else blocks runs only when all the conditions above are False.



```python

if condition:
    # code to be executed
else:
    # code to be executed if all the conditions above are False
```

````python
# same as before, converting the typed text into a real number so we can compare it
age = int(input("Enter your age : "))

if age >= 18:
    # this only runs when the condition above is True
    print("Eligible to drive")
else:
    # "else" is like the backup plan, it catches every other case where the if was False
    # you don't write a condition for else, it just means "anything that didn't match above"
    print("Not Eligible to drive")

# since we entered 12, age >= 18 is False, so python skips the if block entirely
# and jumps straight down to the else block instead, that's why "Not Eligible to drive" got printed
````
