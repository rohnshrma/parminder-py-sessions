# Object-Oriented Programming (OOP) in Python

**In one line:** OOP means bundling *data* and the *actions on that data* into one reusable package called an **object**.

**Real-life picture:** A *blueprint* of a house is a **class**. Each actual house built from it is an **object**. One blueprint → many houses.

| Term | Everyday meaning | Example |
|---|---|---|
| **Class** | Blueprint / template | `Shoe` |
| **Object / Instance** | A real thing made from the blueprint | `s1 = Shoe()` |
| **Attribute / Property** | A fact about the thing | `name`, `size`, `color` |
| **Method** | Something the thing can do | `info()` |
| **`self`** | "This very object" | `self.name` |

## 1. Static Class
A class where every object gets the **same fixed values**. Like a factory that only makes one shoe model.

````python
# class = keyword to create a blueprint; Shoe = its name (capitalised by convention)
class Shoe:
    # properties: values stored in the class; every Shoe starts with these
    name = "Asics 1234"   # shoe's model name
    size = "UK 8"         # shoe's size
    color = "Red"         # shoe's color

    # method: a function inside a class; it describes what a Shoe can do
    # self = the particular shoe that is calling this method
    def info(self):
        print(self)  # prints the object itself (its type and memory address)
        # \n = new line; self.name reads this shoe's name
        print(f"Name : {self.name}\nSize : {self.size}\nColor : {self.color}")
````

### Creating an object
Calling the class like a function, `Shoe()`, builds a real shoe from the blueprint.

````python
s1 = Shoe()  # make an object (instance) of Shoe and store it in s1
````

````python
s1  # shows what s1 is: a Shoe object and where it lives in memory
````

````python
print(type(s1))  # type() tells what kind of thing s1 is -> our Shoe class
````

````python
print(type(123))  # even plain numbers are objects of built-in classes (int)
````

### Using properties and methods
Use a **dot** (`.`) to reach inside an object: `object.property` or `object.method()`.

````python
print(s1.name)   # read the name property
print(s1.size)   # read the size property
print(s1.color)  # read the color property
````

````python
s1.info()  # call the method; Python passes s1 in as "self" automatically
````

````python
# s1 => object / instance
# size, color, name => properties / attributes (data)
# info => method (action)
````

### A second object
Same blueprint, another shoe. Both have identical values here because the class is *static*.

````python
s2 = Shoe()        # a brand-new, separate Shoe object
print(s2)          # different memory address from s1 -> a different object
print(s2.name)
print(s2.size)
print(s2.color)
````

````python
s2.info()  # same method works on any Shoe
````

## 2. Semi-Dynamic Class
Values are **not fixed** in the class; we hand them over *after* creating the object using a method.
Downside: we must remember to call that method, or the values will not exist.

````python
class Person:
    # take_values: a method that receives data and stores it on the object
    # name, age, gender are inputs given by the user
    def take_values(self, name, age, gender):
        self.name = name        # save name inside this person
        self.age = age          # save age inside this person
        self.gender = gender    # save gender inside this person

    # info: prints a sentence using the stored values
    def info(self):
        print(f"{self.name} is a {self.age} old {self.gender}.")
````

````python
parminder = Person()  # empty person: no name/age/gender stored yet
````

````python
parminder.take_values("parminder", 27, "male")  # now fill in the details
````

````python
parminder.info()  # prints: parminder is a 27 old male.
````

## 3. Dynamic Class (with `__init__`)
The best way: values are given **at the moment of creation**, so an object is never half-empty.

**Key ideas**
- **`__init__`** (the *constructor*): runs **automatically** when you create an object. Like filling a form when you join a gym.
- **Dunder methods** (*d*ouble *under*score, e.g. `__str__`): special methods Python calls for you behind the scenes.
- **`__str__`**: decides what `print(object)` shows.
- **`__len__`**: decides what `len(object)` returns.

````python
class Person:
    # __init__ : special / dunder : constructor : runs auto when an instance is created
    def __init__(self, name, age, gender):
        self.name = name                     # store name
        self.age = age                       # store age
        self.gender = gender                 # store gender
        self.expected_life = 90 - self.age   # calculated property: years left, assuming 90
        print("Person created")              # proof that __init__ ran by itself

    # __str__ : runs when we print(object); must return a string
    def __str__(self):
        return f"{self.name} is a {self.age} old {self.gender}."

    # __len__ : runs when we call len(object); here we define "length" as letters in the name
    def __len__(self):
        return len(self.name)
````

````python
rohan = Person("ROhan", 28, "male")  # values go straight to __init__; prints "Person created"
````

````python
rohan.expected_life  # 90 - 28 = 62, calculated for us in __init__
````

````python
print(rohan)  # with __str__ -> a readable sentence instead of a memory address
````

### `len()` on built-ins vs our own object
Because we wrote `__len__`, our object works with `len()` just like a string does.

````python
txt = "hello"  # a normal string (also an object of the str class)
````

````python
print(type(txt))  # <class 'str'>
````

````python
print(len(txt))   # 5 -> str has its own built-in __len__
````

````python
print(len(rohan))  # 5 -> uses our __len__ (letters in "ROhan")
````

## Quick Recap
1. **Class** = blueprint, **object** = thing built from it.
2. **Attributes** = data, **methods** = actions, **`self`** = "this object".
3. **Static** class → same values for all. **Semi-dynamic** → fill values later. **Dynamic** (`__init__`) → fill values at creation (best).
4. **Dunder methods** (`__init__`, `__str__`, `__len__`) let your objects behave like Python's built-in types.

# Inheritance
**In one line:** Inheritance lets a new class *reuse* everything from an existing class, so you don't write the same code twice.

**Real-life picture:** A child inherits traits from their parent, but can also have their own. A `Student` is a `Person`, so a student can do everything a person can, plus more.

| Term | Everyday meaning |
|---|---|
| **Parent / Base / Super class** | The class being inherited from |
| **Child / Derived / Sub class** | The new class that inherits |
| **Overriding** | Child rewrites a parent's method its own way |
| **`super()`** | "Ask my parent to do its part first" |

```python
class Parent:
    pass

class Child(Parent):   # Parent in brackets = inherit from it
    pass
```

## 1. Basic Inheritance
The child gets all the parent's methods for free.

````python
# parent class: holds the common ability
class Person:
    # method: every Person (and anything inheriting from Person) can speak
    def speak(self):
        print("Person can speak")  # message shown when speak() is called
````

````python
# Student(Person) = Student inherits from Person
# pass = empty body; Student adds nothing of its own yet
class Student(Person):
    pass
````

````python
s1 = Student()  # create a Student object
````

````python
s1.speak()  # Student has no speak(), so Python finds it in Person and runs it
````

## 2. Adding Child-Specific Features
A child keeps everything from the parent and can add its own extras.

````python
# redefine Student, still inheriting from Person
class Student(Person):
    # new method that only Student has (Person does not)
    def study(self):
        print("Student is studying")  # message for study()
````

````python
s2 = Student()  # new Student object using the updated class
````

````python
s2.speak()  # inherited from Person
````

````python
s2.study()  # belongs to Student itself
````

## 3. `__init__` and `super()`
If a child has its own `__init__`, the parent's `__init__` is **not** run automatically. `super()` lets the child call it, so the parent's setup isn't repeated.

````python
# parent class
class Book:
    # constructor: saves the title and number of pages
    def __init__(self, title, pages):
        self.title = title    # store title
        self.pages = pages    # store pages

    # prints the book's title
    def info(self):
        print(f"Book Title is {self.title}")
````

````python
# EBook is a Book with an extra property: genre
class EBook(Book):
    # constructor takes everything Book needs, plus genre
    def __init__(self, title, genre, pages):
        super().__init__(title, pages)  # let Book.__init__ store title and pages
        self.genre = genre              # EBook's own extra property
````

````python
# create an EBook; title/pages go to Book's __init__ via super(), genre is saved here
hp = EBook("Harry Potter and the Goblet of Fire", "Fantasy", 1000)
````

````python
hp.pages  # 1000 -> stored by the parent's __init__, still available on the child
````

````python
hp.info()  # inherited method from Book
````

## 4. Method Overriding
If the child defines a method with the **same name** as the parent's, the child's version wins. Like a child doing a chore their own way.

````python
# parent class with a general sound
class Animal:
    def sound(self):
        print("Animal makes a sound")  # generic message
````

````python
# Dog overrides sound() with its own version
class Dog(Animal):
    def sound(self):
        print("Dog Says Woof!")  # dog-specific message
````

````python
a = Animal()  # plain animal
d = Dog()     # dog (a child of Animal)
````

````python
a.sound()  # uses Animal's version -> "Animal makes a sound"
````

````python
d.sound()  # uses Dog's version -> "Dog Says Woof!"
````

# Types of Inheritance
| Type | Plain meaning | Shape |
|---|---|---|
| **Single** | One child, one parent | A → B |
| **Multi-level** | Grandparent → parent → child | A → B → C |
| **Multiple** | One child, two parents | A, B → C |
| **Hierarchical** | One parent, many children | A → B, C |
| **Hybrid** | A mix of the above | diamond shape |

```
SINGLE         MULTI-LEVEL     MULTIPLE        HIERARCHICAL     HYBRID
  A                A            A     B              A             Employee
  ↓                ↓             \   /             / \            /      \
  B                B               C              B   C      Developer  Manager
                   ↓                                              \      /
                   C                                              TeamLead
```

### Single Inheritance
One parent, one child.

````python
# parent
class Employee:
    def show_employee(self):
        print("I am an employee")  # parent's ability

# child: inherits Employee and adds write_code
class Developer(Employee):
    def write_code(self):
        print("I write code")  # child's own ability
````

````python
d = Developer()       # create a developer
d.show_employee()     # inherited from Employee
d.write_code()        # Developer's own method
````

### Multiple Inheritance
One child with **two or more parents**; it gets abilities from all of them.

````python
# first parent
class Teacher:
    def teach(self):
        print("I can teach")

# second parent
class Developer:
    def code(self):
        print("I can code")

# Instructor inherits from both parents (listed in brackets, separated by comma)
class Instructor(Teacher, Developer):
    def train(self):
        print("I can train students")  # Instructor's own method
````

````python
i = Instructor()  # create an instructor
i.teach()         # from Teacher
i.code()          # from Developer
i.train()         # own method
````

### Multi-level Inheritance
A chain: grandparent → parent → child. The last child gets everything above it.

````python
# top of the chain
class Person:
    def speak(self):
        print("Person can speak")

# middle: inherits Person
class Employee(Person):
    def work(self):
        print("I am an employee")

# bottom: inherits Employee (and so Person too)
class Developer(Employee):
    def code(self):
        print("I can code")
````

````python
d = Developer()
d.speak()  # from Person (grandparent)
d.work()   # from Employee (parent)
d.code()   # own method
````

### Hierarchical Inheritance
One parent, **many children**. Each child shares the parent's abilities but has its own skills.

````python
# single parent
class Employee:
    def work(self):
        print("I am an employee")

# child 1
class Developer(Employee):
    def code(self):
        print("i write code")

# child 2
class Designer(Employee):
    def design(self):
        print("i create design")
````

````python
d = Developer()    # developer object
des = Designer()   # designer object
````

````python
d.work()      # shared, from Employee
d.code()      # Developer only
des.design()  # Designer only
des.work()    # shared, from Employee
````

### Hybrid Inheritance
A combination of two or more types. Below: `TeamLead` inherits from `Developer` and `Manager`, which both inherit from `Employee` (a diamond).

````python
# top parent
class Employee:
    def show_employee(self):
        print("I am an employee")

# child of Employee
class Developer(Employee):
    def code(self):
        print("i write code")

# another child of Employee
class Manager(Employee):
    def manage(self):
        print("i manage things")

# inherits from both Developer and Manager -> multiple + hierarchical = hybrid
class TeamLead(Developer, Manager):
    def lead(self):
        print("I lead the development team")
````

````python
tl = TeamLead()      # create a team lead
tl.show_employee()   # from Employee (reached through Developer/Manager)
tl.code()            # from Developer
tl.manage()          # from Manager
tl.lead()            # own method
````

## Quick Recap: Inheritance
1. **Inheritance** = a child class reuses a parent class's code.
2. A child can **add** new methods and **override** parent ones.
3. **`super()`** calls the parent's version (commonly in `__init__`).
4. Types: **single, multi-level, multiple, hierarchical, hybrid**.
