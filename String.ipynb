## How to check data type of a value ?
`type()` function return the data type of value / variable specified

````python
# type() is like asking python "hey, what kind of thing is this?"
x = 23
print(type(x))       # x is a variable holding a whole number -> int
print(type(100))     # 100 is written directly as a whole number -> int
print(type("100"))   # "100" is inside quotes, so python treats it as text, not a number -> str
````

## Note : the data type of a value taken as an input using the `input()` function is of type `Str` by default but can be converted into supported data type


````python
# whatever the user types into input(), python ALWAYS hands it back as text (str)
# even if the user types numbers like 23, python still sees it as the text "23"
name = input("Enter your name : ")
print(name , type(name))   # str, as expected - name is meant to be text

age = input("Enter your age : ")
print(age , type(age))     # still str! even though 23 looks like a number, it's stored as text

# to actually use it as a number (like doing math with it), we must convert it ourselves
# int() converts text that looks like a whole number into a real number
yourAge = int(input("Enter your age : "))
print(yourAge , type(yourAge))   # now it's a real int, ready for math
````

# indexing
counting starting from 0 used to access characters of a string 

`str[start(inclusive)? : stop(exclusive)? : step?(default 1)]` are used to use indexing 



```python

# h e l l o : 
# 1 2 3 4 5 : normal counting (count length)
# 0 1 2 3 4 : positive indexing (L -> R)
#-5-4-3-2-1 : nagative indexing  ( R-> L) 
```

--- 
- first character is always on the 0th position (positive)
- last character is always on the len(str) - 1 = th position (positive)

- first character is always on the -1  position (negative)

````python
# think of a string as a row of numbered boxes, one letter per box - that's what "indexing" means
name = "john donny"
````

````python
# [0] always grabs the very first character, since counting boxes starts at 0, not 1
print(name[0])                    # j - the first letter

# len(name) - 1 gives the position of the LAST character
# (we subtract 1 because counting starts at 0, so the last box is one less than the total count)
print(name[len(name) - 1])        # y - the last letter
````

````python
# negative index = counting from the right end instead of the left
# [-1] is a shortcut for "the last character", no need to calculate len(name) - 1
print(name[-1])
````

````python
# slicing = grabbing a chunk of the string instead of just one letter
# name[start:stop] means "start at this box, and stop right BEFORE this box"
print(name[0:4]) # box 0 up to (but not including) box 4 -> "john"
````

````python
# python is forgiving with slicing - even if the stop number goes past the end of the string
# it just grabs whatever is there and stops, instead of throwing an error like [10] alone would
print(name[5 : 10]) # "fake indexing" - asking for more than exists, python just gives what's available -> "donny"
````

````python
# leaving the "stop" part empty means "go all the way to the end"
print(name[5 : ]) # start at box 5, go to the end -> "donny"
````

````python
# leaving the "start" part empty means "start from the very beginning"
print(name[ : 4]) # from the start up to (not including) box 4 -> "john"
````

````python
# negative numbers work in slices too - "start 5 characters from the end, go to the end"
print(name[-5 : ]) # -5 : end -> "donny"
````

````python
# the 3rd number in a slice is the "step" - how many boxes to jump each time
print(name[ : 4 : 1]) # step of 1 = grab every single character (same as not writing a step at all) -> "john"
print(name[ : 4 : 2]) # step of 2 = grab every OTHER character (skip one each time) -> "jh"
````

````python
# reverse a string
# a step of -1 means "walk backwards through the boxes, one at a time" -> instantly reverses the string
print(name [::-1])
````

````python
# a "palindrome" is a word that reads the same forwards and backwards, like "racecar"
# here we reverse "racecar" with [::-1] and check if it's equal (==) to the original - it is, so True
print("racecar" == "racecar"[::-1])
````

### String Methods


````python
# just checking what's currently stored inside 'name' before we start using string methods on it
name
````

````python
# "string methods" are built-in tools that come attached to every string
# you use them by writing stringVariable.methodName() - the dot is how you "unlock" the tool

print(name.upper())        # turns every letter to UPPERCASE -> "JOHN DONNY"
print(name.lower())        # turns every letter to lowercase -> "john donny"
print(name.title())        # capitalizes the first letter of each word -> "John Donny"
print(name.capitalize())   # capitalizes only the very first letter -> "John donny"

print(name.replace("donny", "smith"))   # swaps one piece of text for another -> "john smith"
print(name.split())                     # breaks the string into a list of words, splitting on spaces -> ['john', 'donny']

print(name.strip())        # removes extra spaces from the start and end (handy for cleaning up input())
print(name.find("donny"))  # tells you the starting box (index) where "donny" begins, or -1 if not found
print(name.count("n"))     # counts how many times "n" shows up in the string

print(name.startswith("john"))  # True/False - does the string begin with this text?
print(name.endswith("donny"))   # True/False - does the string end with this text?

print(name.isalpha())      # True only if EVERY character is a letter (spaces make this False)
print("12345".isdigit())   # True only if EVERY character is a digit
````

### More String Methods
the rest of python's built-in string toolbox, for full coverage

````python
# .join(list) is like the reverse of .split() - it glues a LIST of strings together into ONE string
# you call it on the "glue" (separator) and pass in the list you want stuck together
words = ["hello", "world", "my", "name", "is", "john"]
print("-".join(words))   # glue with a dash -> "hello-world-my-name-is-john"
print(" ".join(words))   # glue with a space -> back to a normal sentence
````

````python
# .strip() removes whitespace from BOTH ends. .lstrip() only cleans the LEFT (start) side, leaving the right side untouched
padded = "   hello world   "
print(padded.lstrip() + "|")   # left space is gone, right space is still there right before the |
````

````python
# .rstrip() is the mirror image of .lstrip() - it only cleans the RIGHT (end) side
print(padded.rstrip() + "|")   # right space is gone, left space is still there before the "hello"
print(padded.strip() + "|")    # .strip() cleans BOTH sides at once - the one we already used earlier
````

````python
# .removeprefix(text) chops that exact text off the START of the string, only if it's actually there (mirror of .removesuffix())
filename = "IMG_holiday.png"
print(filename.removeprefix("IMG_"))   # really does start with "IMG_", so it gets removed -> "holiday.png"
print(filename.removeprefix("DSC_"))   # doesn't start with "DSC_", so nothing changes - string comes back untouched
````

````python
# .ljust(width) pads the string with spaces on the RIGHT until it reaches the given total width (keeps the text left-aligned)
print("cat".ljust(10) + "|")   # "cat" + enough spaces to fill up to 10 characters total, then the |
````

````python
# .rjust(width) pads the string with spaces on the LEFT instead, so the text ends up right-aligned
print("cat".rjust(10) + "|")   # spaces first, then "cat", 10 characters total before the |
````

````python
# .center(width) pads spaces on BOTH sides, trying to center the text within the given width
print("cat".center(10) + "|")   # spaces split roughly evenly on the left and right of "cat"
````

````python
# .zfill(width) is like .rjust(), but it pads with ZEROS instead of spaces - handy for invoice numbers, dates, IDs, etc.
print("7".zfill(3))     # -> "007"
print("42".zfill(5))    # -> "00042"
````

````python
# .partition(separator) splits the string into EXACTLY 3 pieces: (before the separator, the separator itself, after it)
# it only ever splits on the FIRST occurrence, no matter how many times the separator shows up
email = "john.doe@email.com"
print(email.partition("@"))   # -> ('john.doe', '@', 'email.com')
````

````python
# .rpartition(separator) does the same 3-way split, but looks for the LAST occurrence of the separator instead of the first
path = "folder/subfolder/file.txt"
print(path.rpartition("/"))   # -> ('folder/subfolder', '/', 'file.txt')
````

````python
# .rindex(substring) is like .index(), but searches backwards - gives the position of the LAST occurrence
# (basically .rfind()'s stricter twin: .rindex() crashes with an error if the text isn't found, instead of returning -1)
message = "hello world my name is john doe and i am an alien"
print(message.rindex("o"))
````

````python
# .istitle() checks if the string is ALREADY in "title case" - meaning every word starts with a capital letter
print("John Doe".istitle())   # True - every word starts capital
print("John doe".istitle())   # False - "doe" doesn't start with a capital letter
````

````python
# .isspace() -> True only if the string has at least one character, and EVERY character is whitespace (spaces, tabs, newlines)
print("   ".isspace())    # all spaces -> True
print("  a ".isspace())   # has a real letter hiding in there -> False
````

````python
# .isdecimal() is the STRICTEST of the 3 "number" checks (isdecimal / isdigit / isnumeric from earlier)
# True only for plain base-10 digit characters (0-9, and their look-alikes in some other scripts) - no fractions, no superscripts
print("123".isdecimal())   # True - plain digits
print("½".isdecimal())     # False - a fraction symbol, too "fancy" for isdecimal()
print("½".isnumeric())     # True - isnumeric() is the most generous of the three, so it accepts this
````

````python
# .isascii() -> True if EVERY character fits inside plain ASCII (ordinary english letters, digits, common keyboard symbols)
print("hello123".isascii())   # True - totally standard characters
print("héllo".isascii())      # False - the é is outside plain ASCII
````

````python
# .isprintable() -> True if the string has no "invisible/control" characters (like \n or \t) that can't be displayed normally
print("hello world".isprintable())    # True - totally normal, printable text
print("hello\nworld".isprintable())   # False - contains a newline, which counts as non-printable
````

````python
# .format() is the OLDER way of building a formatted string (f-strings, used earlier in this notebook, are the newer/easier way)
# it also uses { } as placeholders, but you fill them in by passing values into .format() instead of writing them inline
print("Hello, my name is {} and I am {} years old".format("John", 25))
print("Hello, my name is {n} and I am {a} years old".format(n="John", a=25))   # placeholders can be named too
````

````python
# .format_map() works like .format(), but takes a DICTIONARY directly instead of separate named arguments
# handy when the values you want to fill in are already sitting together in a dict
info = {"n": "John", "a": 25}
print("Hello, my name is {n} and I am {a} years old".format_map(info))
````

````python
# .casefold() is like .lower(), but even MORE aggressive - built for comparing text across different languages/alphabets
# for plain english text, .lower() and .casefold() behave the same way
print("HELLO".casefold())   # -> "hello"
````

````python
# .encode() converts a string into "bytes" - the raw 0s-and-1s format computers actually use to store/send text
# (matters when saving files or sending text over a network - not something you'll need every single day as a beginner)
print("hello".encode())   # -> b'hello' (the "b" means this is now a bytes object, not a normal string)
````

````python
# .expandtabs() replaces any tab characters (\t) in the string with spaces, lining things up to a fixed width (default 8)
print("a\tb\tc")                # \t shows as a jump to the next "tab stop"
print("a\tb\tc".expandtabs(4))  # forces each tab stop to be exactly 4 spaces wide instead
````

````python
# .translate() swaps out characters based on a "translation table" - .maketrans() is the helper that BUILDS that table
# together they let you replace many different characters all in one go (way faster than chaining lots of .replace() calls)
table = str.maketrans("aeiou", "12345")   # map each vowel to a matching digit
print("hello world".translate(table))     # every a/e/i/o/u gets swapped for its matching digit
````
