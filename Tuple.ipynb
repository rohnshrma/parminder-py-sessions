# Tuple

- tuple can be created using with or without `( )`
- ( ) doesn't ensure something to be a tuple
- the moment you try to put multiple values in a single variable divided by `,` it becomes a tuple
- creating a tuple with a single item requires a `,` after the item

```python
# you don't even need round brackets () to make a tuple
# the comma is the real thing that makes it a tuple, not the brackets
marks = 12,23,34,12,12
print(marks , type(marks))   # prints as (12, 23, 34, 12, 12) - python adds the brackets when showing it, type is tuple

# this is NOT a tuple - there's no comma inside, so the () here is just normal brackets
# python just sees "john" wrapped in brackets, no different than writing "john" alone
names = ("john")
print(names , type(names))   # type comes out as str, not tuple - proves brackets alone don't make a tuple

# to make a tuple with only ONE item, you must put a comma after it, even with no second value
# without the trailing comma this would just be a plain string like the "names" example above
veggies = "Potato",
print(veggies , type(veggies))   # this one really is a tuple: ('Potato',)
```

```python
print(marks) # ordered and allows duplicates
# "ordered" means the items stay exactly in the sequence you typed them, python won't shuffle them
# "allows duplicates" means the same value (like 12) can show up more than once, no complaints
```

```python
print(marks[2]) # indexed
# unlike a dictionary, a tuple DOES have position numbers, starting from 0
# so marks[2] means "give me the item sitting at position 2" which is the 3rd item (34)
```

```python
# tuples are immutable, meaning once created you cannot change, add, or remove an item directly
# trying to overwrite position 2 like you would with a list crashes with a TypeError
marks[2] = 100
```

```python
# since tuples can't be edited directly, the trick to "change" one is:
# 1) convert it into a list (lists CAN be edited)
# 2) make your changes on the list
# 3) convert it back into a tuple
marksList = list(marks)     # turn the tuple into a list so we can modify it
marksList.append(1000)      # add 1000 to the end of the list
marksList.pop(2)            # remove whatever is sitting at position 2 of the list
marks = tuple(marksList)    # convert the edited list back into a tuple and save it over the old "marks"
```

```python
# checking the result of the list-convert-edit-convert-back trick from above
# old value 34 (which was at position 2) is gone, and 1000 got added to the end
marks
```

```python
# returns number of occurances of the item specied
# .count() looks through the whole tuple and tells you how many times that value shows up
marks.count(12)
```

```python
# .index() finds the position of the FIRST time this value appears in the tuple
# 12 shows up first at position 0, so that's what gets returned
marks.index(12)
```

```python
marks.index(12 , 1) # search item, start position (inclusive)
# this time we tell it to skip position 0 and start looking from position 1 onward
# so it finds the NEXT 12 after the one at position 0, which is at position 2
```

```python
# this is called "unpacking" - splitting a tuple's items straight into separate variables in one line
# marks has 5 items, so we need exactly 5 variable names on the left to catch them all, in order
a,b,c,d,e  = marks
```

```python
# a caught whatever was at position 0 in marks
a
```

```python
# b caught whatever was at position 1 in marks
b
```

```python
# c caught whatever was at position 2 in marks
c
```

```python
# the * (star) means "grab everything left over and dump it into this variable as a list"
# a takes the first item, c takes the last item
# b becomes a list holding whatever is left in the middle, however many items that turns out to be
# this is handy when you don't know exactly how many items are in the middle
a,*b,c = marks
```

```python
# a is still just the first item, same as before
a 
```

```python
# b is a LIST now (not a tuple), holding every item that wasn't grabbed by a or c
# notice the square brackets in the output - that's the giveaway it's a list, thanks to the *
b
```

```python
# c grabbed the very last item in marks
c
```
