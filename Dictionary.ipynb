# Dictionary
{ } are used to create a dicitonary where data is stored in a `key:value` pair format also known as a property (divided by `,`)

---
Ordered, Unindexed, Changable(mutable), No duplicates are allowed

```python
# curly braces {} make a dictionary
# think of it like a real dictionary - you look up a "word" (the key) to get its "meaning" (the value)
# each key:value pair is separated by a comma
car = {
    "model" : "i20 asta",       # key is "model", value is "i20 asta"
    "brand" : "hyundai",        # key is "brand", value is "hyundai"
    "year" : 2020,               # values don't have to be text, can be a number too
    "availableColors" : ["red","black"],   # a value can even be a list
    "model" : "Verna"            # careful - "model" key already used above!
                                  # dictionaries don't allow duplicate keys
                                  # so this second "model" just overwrites the first one
}
```

```python
print(type(car))   # tells us what kind of thing "car" is - here it prints <class 'dict'>
print(len(car))    # counts how many key:value pairs are inside - notice it says 4, not 5
                    # that's proof the duplicate "model" key from before got merged into one
print(car) #  ordered and no duplicates allowed
# "ordered" means python remembers the order you typed the keys in and will print them back in that same order
# "no duplicates" means if you repeat a key, the last value you gave it wins, old one is gone
```

```python
print(car["model"]) # unindexed
# with a list you'd grab something using a number position like car[0]
# but a dictionary has no positions/index numbers, you grab things by their key name instead
# so car["model"] means "give me whatever value is stored under the key called model"
```

```python
car["year"] = 2022 # changable
# this line says "go to the key called year, and replace whatever is there with 2022"
# this proves dictionaries are mutable/changable - you can edit values after creating the dictionary
print(car)
```

```python
# keys don't have to be words, they can be numbers too
# this looks a bit like a list because the keys happen to be 0 and 1
# but it's still a real dictionary, python doesn't automatically know it "should" be a list
names = {
    0 : "john",
    1 : "james"
}
```

```python
# grabbing the value stored at key 0, this returns "john"
# don't get confused, this "0" is a key name here, not a list index
names[0]
```

```python
# key 2 doesn't exist yet in "names"
# when you assign to a key that isn't there, python doesn't throw an error
# instead it just creates that new key:value pair on the spot
names[2] = "Jackie"
```

```python
# confirming the new key 2 with value "Jackie" got added to the dictionary
print(names)
```

```python
# .get() is another way to read a value using its key, works the same as names[0] here
# the difference between .get() and square brackets shows up when the key is missing - see next cells
names.get(0)
```

```python
print(names[5]) # key error if key not found
# there is no key 5 in "names", so square bracket access crashes the program with a KeyError
# this is the risky way to read a dictionary - only use it when you're 100% sure the key exists
```

```python
print(names.get(5)) # returns none  if key not found
# .get() is the safe way to read a dictionary - if the key isn't there it just gives back None
# instead of crashing your whole program, this is why .get() is usually the better habit to use
```

```python
# .update() lets you change/add many key:value pairs in one go, instead of doing it one line at a time
# pass it another dictionary - if a key already exists its value gets overwritten
# if a key doesn't exist yet, it gets added as a brand new pair
names.update({0 : "Parminder" , 10 : "Sunny"})
# key 0 already existed ("john") so it gets replaced with "Parminder"
# key 10 is new so it gets added fresh
print(names)
```

```python
# returns a list of keys
# .keys() gives you every key in the dictionary, packed inside a special "dict_keys" view
# you can loop over it just like a list if you need to check what keys exist
names.keys()
```

```python
# return list of values
# .values() is the sister of .keys() - instead of key names it gives you every value stored
names.values()
```

```python
# returns a list of properties where each property is in a form of tuple
# .items() gives you both the key AND the value together, paired up as (key, value) tuples
# this is super handy when you want to loop through a dictionary and use both the key and value at once
names.items()
```

```python
# removes the property from the key specified, raises a key error if key  not found
# .pop() deletes that key:value pair from the dictionary AND hands you back the value that was removed
names.pop(1) # key name
# so this removes key 1 (james) from names, and the cell shows 'james' because that's what got returned
```

```python
# key 100 doesn't exist in names, so .pop() has nothing to remove
# just like plain square brackets, this crashes with a KeyError
# tip: .pop(100, "not found") lets you give a default instead of crashing, same safety idea as .get()
names.pop(100)
```

```python
# removes the last key value pair and it returns it as a tuple
# .popitem() doesn't need a key at all - it just yanks out whichever pair was added most recently
# handy when you just want to "undo" the last thing you added without remembering its key
names.popitem()
```

```python
# dict.fromkeys() is a shortcut to build a dictionary when every key should start with the SAME value
# first argument is a list of the keys you want, second argument is the one value all of them will share
# note: this only works because the "x = 34" cell below was already run before this one
# (in a notebook, cells run in whatever order YOU click run, not top to bottom order on the page)
stu = dict.fromkeys( ["Rohan" , "Parminder" , "Sukhjeet"] , x )
```

```python
# check the result - all three names got the same value 34
# handy for setting a "starting score" or default for a bunch of people/items at once
stu
```

```python
# a small fresh dictionary to demo .setdefault() next
d = {"a" : 1}
```

```python
# .setdefault(key, value) means: "if this key already exists, just give me its current value and do nothing else"
# key "a" already exists with value 1, so it just returns the existing 1
# the 100 we passed gets ignored completely because it was NOT needed
d.setdefault("a" , 100)
```

```python
# now key "b" does NOT exist yet in d
# so this time .setdefault() actually creates it, sets its value to 100, and returns that 100
# this is the real use case for setdefault: "add this key only if it's missing, otherwise leave it alone"
d.setdefault("b" , 100)
```
