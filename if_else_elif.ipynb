# if
the code inside of the if blocks runs only when the given condition is True



```python

if condition:
    # code to be executed

```

````python
# input() always gives back text (a string), so we wrap it in int() to turn it into a whole number
# otherwise we can't compare it with >= like a number
age = int(input("Enter your age : "))

# this is the actual "if" check, python reads it like a question: "is age 18 or more?"
# the colon at the end just means "here comes the block of code that runs if this is true"
if age >= 18:
    # this line is indented (pushed in with spaces), and that indentation is what tells python
    # "this line belongs to the if statement above", it will ONLY run when the condition is True
    print("Eligible to drive")

# note: since we typed 12 above, 12 >= 18 is False, so the print line never ran, nothing got printed
````

# else
the code inside of the else blocks runs only when all the conditions above are False.



```python

if condition:
    # code to be executed
else:
    # code to be executed if all the conditions above are False
```

````python
# same as before, converting the typed text into a real number so we can compare it
age = int(input("Enter your age : "))

if age >= 18:
    # this only runs when the condition above is True
    print("Eligible to drive")
else:
    # "else" is like the backup plan, it catches every other case where the if was False
    # you don't write a condition for else, it just means "anything that didn't match above"
    print("Not Eligible to drive")

# since we entered 12, age >= 18 is False, so python skips the if block entirely
# and jumps straight down to the else block instead, that's why "Not Eligible to drive" got printed
````

# Elif (else if)
a fallback condition which comes after if or elif

think of it like a checklist: python checks the `if` first, and only if that's False does it even
bother looking at the first `elif`. if that's also False, it checks the next `elif`, and so on.
the moment ONE of them turns out True, python runs just that one block and skips every condition
below it, even if some of those would also have been True. `else` at the very end (if you add one)
is the "none of the above" catch-all, it has no condition of its own.

````python
# if else statement to handle 5 cases of age
age = int(input('enter age:'))  # turn the typed text into a number so we can use < and >

if age > 0 and age <= 4:
    # "and" means BOTH sides have to be True at once
    # this only runs when age is more than 0, AND age is 4 or less - basically ages 1 to 4
    print('free')
elif age >= 5 and age <= 10:
    # python only checks this because the line above was False
    # covers ages 5 through 10
    print('$5')
elif age >= 11 and age <= 14:
    # covers ages 11 through 14
    print('$10')
elif age >= 15 and age <= 100:
    # covers ages 15 through 100
    print('$12')
else:
    # catches everything that didn't fit any range above: 0, negative numbers, or anything past 100
    print('invalid input')

# this whole block is really just the price chart from the cell above, translated into code:
# 1-4 free, 5-10 $5, 11-14 $10, 15-100 $12, anything outside that range is invalid
````
