```python
# Arithmetic Operators
```

```python
# just storing two numbers in variables so we can use them below
# think of a variable as a labeled box where you keep a value
a = 35 
b = 3
```

```python
print(a + b) # addition : simply adds the two numbers together, 35 + 3 = 38
print(a - b) # subtraction : takes b away from a, 35 - 3 = 32
print(a * b) # multiplication : a repeated b times basically, 35 * 3 = 105
print(a / b) # division : splits a into b equal parts, gives a decimal answer (this is called "true division")
print(a % b) # modulus : divides a by b and tells you only the leftover/remainder, not the answer itself
print(a ** b) # exponent/power : multiplies a by itself b times, so 35 * 35 * 35
print(a // b) # floor division : does normal division but then chops off everything after the decimal point (rounds down)
```

```python
# Assignment Operators
```

```python
x = 23 # regular assignment, we're just putting the number 23 into the box called x
print(x)

# x = x + 10 
# the line above is the long/manual way of doing it, the line below (+=) does the exact same thing but shorter
x += 10  # add and assign : take whatever is in x, add 10 to it, then save that new value back into x -> 23 + 10 = 33
print(x)
x -= 10  # sub and assign : take whatever is in x, subtract 10 from it, save it back -> 33 - 10 = 23
print(x)
x *= 10  # multiply and assign : take x, multiply by 10, save it back -> 23 * 10 = 230
print(x)
x /= 10  # divide and assign : take x, divide by 10, save it back -> 230 / 10 = 23.0 (note it becomes a decimal now)
print(x)
x %= 10  # remainder and assign : divide x by 10 and just keep the leftover, save it back -> 23.0 % 10 = 3.0
print(x)
x **= 10  # power and assign : raise x to the power of 10, save it back -> 3.0 ** 10 = 59049.0
print(x)
x //= 10  # floor divide and assign : divide x by 10 and round down, save it back -> 59049.0 // 10 = 5904.0
print(x)
```

```python
# comparison operators
```

```python
# these all compare a (35) and b (3) and give back True or False, nothing else
print(a == b) # equals to : checks if both are exactly the same value, 35 is not 3, so False
print(a != b) # not equals to : checks if they are different, they are different, so True
print(a > b) # greater than : is a bigger than b? yes, 35 > 3, so True
print(a >= b) # greater than or equal to : is a bigger than or the same as b? yes, so True
print(a < b) # less than : is a smaller than b? no, 35 is not less than 3, so False
print(a <= b) # less than or equal to : is a smaller than or the same as b? no, so False
```

# logical operators
- and : expects all the condition to be True to give a Truthy answer
- or : expects at least one of the condition to be True to give a Truthy answer
- not : reversing the boolean value

```python
# "and" is strict, BOTH sides need to be True for the whole thing to be True
# if even one side is False, the whole thing becomes False
print(12 > 3 and 3 < 12)  # left side: 12 > 3 is True, right side: 3 < 12 is True, True and True = True
print(12 < 3 and 3 < 12)  # left side: 12 < 3 is False, right side is True, but False and True still = False
print(12 < 3 and 3 > 12)  # both sides are False here, False and False = False
```

```python
# "or" is chill, it only needs ONE side to be True for the whole thing to be True
# it only becomes False when both sides fail
print(12 > 3 or 3 < 12)  # both sides happen to be True here, True or True = True
print(12 < 3 or 3 < 12)  # left side is False, but right side is True, so as a whole it's still True
print(12 < 3 or 3 > 12)  # both sides are False, so there's nothing to save it, False or False = False
```

```python
# "not" just flips whatever boolean value you give it, True becomes False and vice versa
print(not True) # True flipped is False
print(not False) # False flipped is True
print(not 12 > 3) # first python checks 12 > 3, that's True, then "not" flips it to False
```
