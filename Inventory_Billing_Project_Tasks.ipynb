## Business rules (refer back to these)

**Product types and GST**

| Type | GST | Extra detail stored |
|---|---|---|
| `general` | 12% | none (store `-`) |
| `perishable` | 5% | days left before expiry |
| `electronic` | 18% | warranty in months |

**Other rules**

- A product is **low on stock** when its stock is below 5.
- A perishable product with **2 or fewer days left** gets 20% off its price (before GST).
- An electronic product has no special discount.
- Bill discount: bill total of 1000 or more gets 5% off, 5000 or more gets 10% off.
- A **premium** customer gets an extra 5% off on top of the bill discount.

**File formats** (one record per line, values separated by commas)

`products.txt`
```
101,Milk,45.0,20,perishable,3
102,Rice 5kg,420.0,15,general,-
103,Earphones,799.0,8,electronic,6
```
Order: product id, name, price, stock, type, extra detail

`customers.txt`
```
C01,Amit,regular
C02,Neha,premium
```
Order: customer id, name, type

`sales.txt` (one line for each item sold)
```
1001,C02,101,Milk,perishable,2,94.5
```
Order: bill number, customer id, product id, product name, type, quantity, line total

`errors.txt` (one line for each error caught)
```
ProductNotFoundError: No product with id 999
```
