# Functions
functions are reusable block of code, where the code inside of a function runs only when called by its name followed by `()`

---
`def` keyword is used to define a runction followed by function name and a pair of `()` having parameters (optional)


```python

# creating a function
def funcName(parameters):
    # code to be executed

# calling a function
funcName()
```

````python
# this version of calcBmi takes NO parameters - it asks the user for weight and height itself, every time it runs
def calcBmi():
    # input() always gives back a string, so we wrap it in int()/float() to turn it into a real number we can do maths on
    w = int(input("Enter weight in kgs : "))
    h = float(input("Enter height in meters : "))

    # this is the actual bmi formula: weight divided by height squared
    # h ** 2 means h to the power of 2 (h * h)
    bmi = w / h ** 2

    # now we just check which bracket the bmi falls into and print the matching message
    if bmi < 18.5:
        print(f"under weight as your bmi is {bmi}")
    elif bmi >= 18.5 and bmi < 25:
        print(f"normal weight as your bmi is {bmi}")
    elif bmi >= 25 and bmi < 30:
        print(f"overweight as your bmi is {bmi}")
    else:
        print(f"obese as your bmi is {bmi}")

# notice this function doesn't return anything - it just prints straight to the screen and ends
# that's fine when you only need to display a result, but it means you can't save/reuse the answer later
# (there's no value to store because nothing was "returned" back to whoever called the function)
````

````python
# calling the function twice - each call is completely separate, python runs the whole
# function body from top to bottom again and asks for fresh input both times
calcBmi()
calcBmi()
````

````python
# this time calcBmi takes w and h as PARAMETERS - values we hand it directly when calling it
# no more input() needed, which makes the function way more reusable (great for looping over
# many people's data later, or testing with fixed numbers)
def calcBmi(w , h):
    bmi = w / h ** 2
    if bmi < 18.5:
        print(f"under weight as your bmi is {bmi}")
    elif bmi >= 18.5 and bmi < 25:
        print(f"normal weight as your bmi is {bmi}")
    elif bmi >= 25 and bmi < 30:
        print(f"overweight as your bmi is {bmi}")
    else:
        print(f"obese as your bmi is {bmi}")

# note: this re-defines calcBmi, so it completely replaces the earlier no-parameter version
# from now on, calling calcBmi() without arguments would error out - python only remembers the latest definition
````

````python
# 100 and 1.80 slot into w and h in that exact order - first value goes to the first parameter, and so on
# this is called "positional" arguments, position decides which value goes where
calcBmi(100 , 1.80)
calcBmi(100 , 1.68)
````

````python
# return doesn't just "send a value back" - it also immediately stops/exits the function right there
# so anything written after a return statement inside that same block never runs
def prints():
    print(1)
    print(2)
    return "hello" # function gets terminated here (code below is unreachable)
    print(3)
    print(4)

# print(3) and print(4) are dead code - python never reaches them because return already ended the function
# this is different from calcBmi above, which used print() only and had no return at all -
# print() just displays something on screen, it does NOT hand a value back to whoever called the function
````

````python
# calling prints() runs it, which prints 1 and 2 to the screen, then returns "hello"
# because it's a return value, we can actually catch it and store it in a variable, here called x
x = prints()

# just running the function alone (without x =) would still print 1 and 2, but the returned
# "hello" would be thrown away since nothing is there to catch it
````

````python
# x now genuinely holds the string "hello" (the returned value), proof that return actually works
# this is the whole point of returning - the value survives after the function finishes and we can use it later
x
````

````python
# a small, focused function - takes one number x, and gives back x squared (x ** 2 means x to the power of 2)
# this is the cleanest kind of function: one job, takes an input, returns an output, nothing printed inside it
def sq(x):
    return x ** 2
````

````python
# calling sq(2) sends 2 in as x, so it returns 2 ** 2 which is 4, and we store that in a variable
# heads up: we're reusing the name x here, so this OVERWRITES the earlier x that held "hello"
# a variable can only hold one value at a time - the old "hello" is gone the moment this line runs
x = sq(2)
````

````python
# confirms x is now 4 - the squared result we got back from sq(), not the "hello" string anymore
x
````

# List Comprehensions
A list comprehension builds a list in a single line. Think of it as a `for` loop + `append` packed inside `[ ]`.

| Shape | What it does |
|---|---|
| `[expr for i in items]` | change/collect every item |
| `[expr for i in items if cond]` | keep only items that pass `cond` |
| `[a if cond else b for i in items]` | pick one of two values for every item |

````python
# list comprehension = a short way to build a new list in ONE line
# instead of: make empty list -> loop -> append, we write it all inside [ ]
# it is just a for loop squeezed into brackets, nothing magical
````

````python
# pattern: [what_to_store   for  item  in  where_items_come_from]
# range(1,11) gives 1,2,3...10 (11 is NOT included, the end is always skipped)
# "i" is the item we grab each round, and the first "i" is what gets stored in the list
# so this just collects 1 to 10 as they are
nums = [ i for i in range(1,11) ]
````

````python
# just typing the variable name at the end of a cell shows what's inside it
nums
````

````python
# this is the exact same thing as the comprehension above, written the long way
# I keep it here so you can compare - same result, way more lines
nums = []                   # start with an empty box
for i in range(1,11):       # go through 1 to 10 one by one
    nums.append(i)          # put each number into the box at the end

print(nums)
````

````python
# the first part doesn't have to be plain "i" - we can do maths on it before storing
# here every number gets squared (i ** 2 means i times i), so we store 1,4,9,16...100
sqs = [ i ** 2 for i in range(1,11) ]
````

````python
sqs
````

````python
# adding an "if" at the end works like a gate - only items that pass get stored
# pattern: [what_to_store  for item in items  if condition]
````

````python
# i % 2 gives the remainder after dividing by 2 - even numbers leave 0, odd numbers leave 1
# so "!= 0" means "not even" = odd. Only odd numbers get through the gate
# note: the if here FILTERS (drops items), it doesn't change them
odds = [ i for i in range(1,11) if i % 2 != 0]
print(odds)
````

````python
# if you want to change what gets stored (instead of dropping items), use if-else
# and notice it goes at the FRONT this time, before the "for"
# pattern: [ value_if_true  if condition  else  value_if_false  for item in items ]
# easy way to remember: "if" at the end = filter, "if-else" at the front = choose between two values
````

````python
# checking nums is still 1 to 10 (we'll use it in the next cell)
nums
````

````python
# for every number in nums we build a text label
# f"{i}-Even" puts the number into the text, so 2 becomes "2-Even"
# every number gets a label - the if-else only decides WHICH label, nobody gets dropped
det = [f"{i}-Even" if i % 2 == 0 else f"{i}-Odd" for i in nums]
````

````python
det
````

# Lambda Functions
A lambda is a small nameless function written in one line. Use it when the function is short and used only once.

````python
# lambda = a tiny function with no name, written in one line
# good for small throwaway jobs where a full def feels like too much
````

````python
# the normal way: def, a name, parameters and return
# this one gives back the bmi instead of printing it, so we can save it in a variable
def calcbmi (w,h):
    return w / h ** 2
````

````python
# pattern: lambda  parameters : what_to_return
# there's no "return" word here - a lambda automatically returns whatever comes after the colon
# it can only hold ONE expression, so no loops or multiple lines inside
````

````python
# same as calcbmi above, but as a lambda
# we store it in the name cbmi so we can call it later like a normal function
cbmi = lambda w ,h : w / h ** 2
````

````python
# 100 goes into w, 1.80 goes into h (positional order, same as normal functions)
x = cbmi(100 , 1.80)
````

````python
# the bmi value that the lambda handed back
x
````

````python
# a list of random numbers to practice with - some odd, some even
ns = [23,45,23,45,5,76,7,8,454545,3324,54434]
````

````python
# the old way to pick only even numbers: empty list, loop, check, append
# 4 lines of work for something a comprehension or filter can do in 1
evens = []
for i in ns:
    if i % 2 == 0:      # even = divides by 2 with nothing left over
        evens.append(i)
````

# filter() and map()
Two built-in helpers that take a function and a list.
- `filter` -> keeps some items
- `map` -> changes every item

````python
# filter picks the items that pass a test and throws the rest away
# filter(function , iterable)
#   function  -> a test that says yes/no for each item
#   iterable  -> the list (or anything loopable) to go through
# it gives back a "filter object" (not a list), so we wrap it in list() to see the values
````

````python
# case 1: using a normal named function as the test
# filter keeps an item if the function's return value counts as True (truthy)
# here we return x itself when it's even, and return nothing (None, which counts as False) when it's odd
# small catch: if x was 0, it would be returned but 0 counts as False too, so 0 would wrongly get dropped
# safer way is to write "return x % 2 == 0" which gives a proper True/False
def evenFilter(x):
    if x % 2 == 0:
        return x
````

````python
# we pass evenFilter WITHOUT brackets - we're handing over the function itself, not calling it
# filter calls it for us on every item in ns
list(filter( evenFilter , ns))
````

````python
# case 2: the test is written as a lambda right inside filter, so no separate def needed
# passing a function into another function like this is called a callback
# lambda x : x % 2 == 0 gives True/False directly, which is the cleanest kind of test
list(filter( lambda x : x % 2 == 0 , ns) )
````

````python
# higher order function = a function that takes another function as input (or gives one back)
# filter and map are both examples - they receive our lambda and use it
````

````python
# map = do the same action on EVERY item and give back all the results
# unlike filter, nothing gets dropped - output has the same number of items as the input
# (also returns a map object, so wrap it in list() to see it)
````

````python
# lambda x : x * 2 doubles each number, map applies it to every item in ns
list(map( lambda x : x * 2 , ns))
````

````python
# quick way to remember the difference:
# filter : asks a yes/no question on each item, keeps the yes ones (result can be SHORTER)
# map : changes each item into something else (result is the SAME length)
````

````python
# filter with the test "is it odd?" on 1 to 20 -> only the odd numbers come out
list(filter( lambda x : x % 2 != 0 , range(1,21) ))
````

````python
# same lambda but with map - watch the difference!
# map doesn't drop anything, it just replaces each number with the answer of the test
# so we get 20 True/False values instead of only the odd numbers
list(map( lambda x : x % 2 != 0 , range(1,21) ))
````
