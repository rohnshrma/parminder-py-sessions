# Functions
functions are reusable block of code, where the code inside of a function runs only when called by its name followed by `()`

---
`def` keyword is used to define a runction followed by function name and a pair of `()` having parameters (optional)


```python

# creating a function
def funcName(parameters):
    # code to be executed

# calling a function
funcName()
```

````python
# this version of calcBmi takes NO parameters - it asks the user for weight and height itself, every time it runs
def calcBmi():
    # input() always gives back a string, so we wrap it in int()/float() to turn it into a real number we can do maths on
    w = int(input("Enter weight in kgs : "))
    h = float(input("Enter height in meters : "))

    # this is the actual bmi formula: weight divided by height squared
    # h ** 2 means h to the power of 2 (h * h)
    bmi = w / h ** 2

    # now we just check which bracket the bmi falls into and print the matching message
    if bmi < 18.5:
        print(f"under weight as your bmi is {bmi}")
    elif bmi >= 18.5 and bmi < 25:
        print(f"normal weight as your bmi is {bmi}")
    elif bmi >= 25 and bmi < 30:
        print(f"overweight as your bmi is {bmi}")
    else:
        print(f"obese as your bmi is {bmi}")

# notice this function doesn't return anything - it just prints straight to the screen and ends
# that's fine when you only need to display a result, but it means you can't save/reuse the answer later
# (there's no value to store because nothing was "returned" back to whoever called the function)
````

````python
# calling the function twice - each call is completely separate, python runs the whole
# function body from top to bottom again and asks for fresh input both times
calcBmi()
calcBmi()
````

````python
# this time calcBmi takes w and h as PARAMETERS - values we hand it directly when calling it
# no more input() needed, which makes the function way more reusable (great for looping over
# many people's data later, or testing with fixed numbers)
def calcBmi(w , h):
    bmi = w / h ** 2
    if bmi < 18.5:
        print(f"under weight as your bmi is {bmi}")
    elif bmi >= 18.5 and bmi < 25:
        print(f"normal weight as your bmi is {bmi}")
    elif bmi >= 25 and bmi < 30:
        print(f"overweight as your bmi is {bmi}")
    else:
        print(f"obese as your bmi is {bmi}")

# note: this re-defines calcBmi, so it completely replaces the earlier no-parameter version
# from now on, calling calcBmi() without arguments would error out - python only remembers the latest definition
````

````python
# 100 and 1.80 slot into w and h in that exact order - first value goes to the first parameter, and so on
# this is called "positional" arguments, position decides which value goes where
calcBmi(100 , 1.80)
calcBmi(100 , 1.68)
````

````python
# return doesn't just "send a value back" - it also immediately stops/exits the function right there
# so anything written after a return statement inside that same block never runs
def prints():
    print(1)
    print(2)
    return "hello" # function gets terminated here (code below is unreachable)
    print(3)
    print(4)

# print(3) and print(4) are dead code - python never reaches them because return already ended the function
# this is different from calcBmi above, which used print() only and had no return at all -
# print() just displays something on screen, it does NOT hand a value back to whoever called the function
````

````python
# calling prints() runs it, which prints 1 and 2 to the screen, then returns "hello"
# because it's a return value, we can actually catch it and store it in a variable, here called x
x = prints()

# just running the function alone (without x =) would still print 1 and 2, but the returned
# "hello" would be thrown away since nothing is there to catch it
````

````python
# x now genuinely holds the string "hello" (the returned value), proof that return actually works
# this is the whole point of returning - the value survives after the function finishes and we can use it later
x
````

````python
# a small, focused function - takes one number x, and gives back x squared (x ** 2 means x to the power of 2)
# this is the cleanest kind of function: one job, takes an input, returns an output, nothing printed inside it
def sq(x):
    return x ** 2
````

````python
# calling sq(2) sends 2 in as x, so it returns 2 ** 2 which is 4, and we store that in a variable
# heads up: we're reusing the name x here, so this OVERWRITES the earlier x that held "hello"
# a variable can only hold one value at a time - the old "hello" is gone the moment this line runs
x = sq(2)
````

````python
# confirms x is now 4 - the squared result we got back from sq(), not the "hello" string anymore
x
````
